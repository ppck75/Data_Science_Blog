## 수업 필기


Python은 데이터 분석에서 가장 널리 사용되는 언어 중 하나이다.  
데이터를 구조화하고, 계산하고, 분석하는 전 과정에서 Python은 높은 유연성을 제공한다.  
이 글에서는 Python 데이터 분석의 기초가 되는 핵심 개념과 문법을 정리한다.

### Numeric Type (숫자형 데이터)

Python의 숫자형 데이터는 대표적으로 `int`와 `float`가 있다.  
정수와 실수는 별도의 선언 없이도 자연스럽게 연산된다.

```python
a = 10        # int
b = 3.14      # float
```

a + b
a * b
연산 과정에서 정수와 실수가 함께 사용되면 결과는 자동으로 실수형으로 변환된다.


### String Type (문자열)
문자열은 텍스트 데이터를 다루는 기본 자료형이다.

text = "Data Analysis"
문자열은 인덱스를 통해 개별 문자에 접근할 수 있으며,
Python의 인덱스는 0부터 시작한다.

```python
text[0]     # 'D'
text[1]     # 'a'
text[-1]    # 's'
```

### List Type (리스트)
리스트는 여러 개의 값을 하나의 객체로 저장할 수 있는 자료형이다.
데이터 분석에서 가장 자주 사용되는 구조 중 하나이다.

numbers = [1, 2, 3, 4, 5]
리스트는 인덱싱뿐 아니라 요소의 추가, 수정, 삭제가 가능하다.

```python
numbers.append(6)     # 요소 추가
numbers[0] = 10       # 요소 수정
numbers.remove(3)     # 요소 삭제
```

### Tuple Type (튜플)
튜플은 리스트와 유사하지만 값을 변경할 수 없는 자료형이다.
변경되면 안 되는 값이나 설정값을 다룰 때 유용하다.

coords = (37.5, 127.0)


### Dictionary Type (딕셔너리)
딕셔너리는 key-value 구조를 가진 자료형이다.
구조화된 정보를 다룰 때 매우 유용하다.

```python
student = {
    "name": "Alice",
    "age": 23,
    "major": "Statistics"
}

student["name"]
student["age"]
```

### Set Type (집합)
집합(set)은 중복을 허용하지 않는 자료형이다.

```python
values = {1, 2, 2, 3}
values
```

중복 제거가 필요할 때 유용하게 활용할 수 있다.


### 조건문 (if / elif / else)  
조건문은 조건에 따라 다른 결과를 수행할 때 사용한다.

```python
score = 85

if score >= 90:
    result = "A"
elif score >= 80:
    result = "B"
else:
    result = "C"
```


### 반복문 (for / while)
반복문은 동일한 작업을 여러 번 수행할 때 사용된다.
데이터 전처리 과정에서 필수적인 문법이다.

```python
for i in range(5):
    print(i)
count = 0
while count < 5:
    print(count)
    count += 1
```

### 함수 (Function)
함수는 반복되는 작업을 정리하고 코드의 재사용성을 높여준다.

```python
def add(x, y):
    return x + y

add(3, 5)
```

분석 과정이 복잡해질수록 함수의 중요성은 더욱 커진다.

*실습을 위해 Pandas, Numpy 라이브러리를 load하였다*

````python
# Create multiple series
series1 = pd.Series(["male","female","female","male","male","male","female"])
series2 = pd.Series([54, 34, 556, 57, 88, 856, 23])
# Combine series into a DataFrame
data = pd.DataFrame({'gender': series1, 'salary': series2}) # column name
print(data)
````

````python
data['salary'].mean() # including 556 and 856 inflates mean value
````

````python
# define the function to recode values above 500 to nan
def recode_salary(value):
    if value > 500:
        return np.nan # recode values above 500 into missing values
    else:
        return value # copy all the other values
````

````python
# apply the function to the 'salary' column
data['salary'] = data['salary'].apply(recode_salary)
````

````python
data['salary'].value_counts(dropna=False).sort_index(ascending=True) # frequency showing missing value as NaN
````

````python
data['salary'].mean() # 556 and 856 are now treated as missing values
````

### saving data WITHOUT missing values

````python
data_filtered=data.dropna() # save data WITHOUT missing values as a new object
````

````python
data_filtered['salary'].value_counts(dropna=False).sort_index(ascending=True) # frequency showing missing value as NaN
````

````python
data_filtered['salary'].mean()
````

### replacing missing values with mean

````python
# you can also replace missing values with the mean of the other values
data['salary'] = data['salary'].replace(np.nan, np.mean(data['salary']))
````
