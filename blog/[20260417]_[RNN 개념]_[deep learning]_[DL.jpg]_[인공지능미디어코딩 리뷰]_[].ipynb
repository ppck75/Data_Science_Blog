# RNN

## 왜 RNN이 필요한가?


시퀀스 데이터: 순서가 의미를 가지는 데이터

일반 신경명(FFNN)은 입력간의 순서를 고려하지 않는다. 그런데, 시퀀스 데이터인 텍스트, 음성, 주가, 날씨 등은 순서가 데이터의 핵심 정보이다

*참고로 FFNN은 입력이 한 방향으로 흘러가는 신경망이다. 즉, 순환하지 않는 신경망으로, input -> hid -> oupput으로 전달된다! 물론, FFNN도 역전파 일어나지만, 여기서 FF라는 것은 학습방법이 아닌 연결구조를 말하는 것이다. 즉, 입력 -> 출력으로 가는 구조!*

*MLP는 FFNN의 한 종류*


그래서 순서를 기억하는 신경망이 필요하다.

FFNN의 한계
> 모든 입력을 독립적으로 처리함, 입력 길이가 고정되어야함. 

그래서 RNN이 등장했다. RNN은 이전 단계의 출력을 다음 단계의 입력으로 전달 즉, '기억'을 가진 신경망이다.


> RNN은 FFNN가 아님! 왜냐하면 출력이 다시 같은 네트워크 안으로 되돌아가는 연결 구조가 있기 때문이다. RNN은 hidden state가 다음 시점 계산에 다시 들어가기 때문.

## RNN (Recurrent Neural Network)

쉽게 말해서 RNN은 우리가 이전에 배운 DL 아키텍처가 단어(또는 토큰이 일반적) 하나하나 마다 구성이 된 것이다. 

이렇게 각 토큰을 순서대로 처리하는데, 우리가 궁금한 것은 각 토큰마다의 결과라기 보다는 문장 전체가 긍정이냐 부정이냐? -> 문장 전체의 분류에 관심이 있다. 그러므로 **각 토큰마다 있는 딥러닝 모델을 연합시여야 한다**  
> 그러므로 RNN은 딥러닝 모델을 연쇄(순환)으로 구성한다

  - 텍스트는 보통 토큰 시퀀스로 들어간다.                                               
  - RNN은 각 시점에서 토큰을 하나씩 처리한다.                                           
  - 각 시점이 완전히 독립적이면 시퀀스 정보를 반영하지 못한다.                          
  - 그래서 이전 정보가 다음 시점 계산에 전달되도록 연결한다.

그런데, 여기서 주의할 것은, 같은 가중치를 각 시점에 반복 적용한다는 것이다. 즉, 시간축으로 쭉 펼쳐보면 여러개의 모델처럼 보이지만? 실제로는 하나의 모델을 반복 사용하는 것이다. (우리가 이해할 때는 여러 상자들을 쭉 나열하지만, 사실 상자 하나 안에서 돌고 도는 것이니.. 같은 가중치가 계속 적용되는 것이다)

즉, 각 토큰마다 별도의 모델이 있는 것처럼 보이지만, 그게 아니라 동일한 RNN 셀과 가중치를 각 시점에 반복 적용한다. 단, 이전 시점의 hidden state를 다음 시점으로 넘겨 문맥을 전달하는 것이다. 

  - 현재 입력 토큰 x_t                                                                                                                                                                             
  - 이전 hidden state h_{t-1}                                                                                                                                                                      
                                                                                                                                                                                                   
  이 두 개를 이용해 현재 hidden state h_t를 계산


결국 RNN의 핵심은 이전 상태를 다음 계산에 재사용하는 '순환구조'에 있다. 

각시점t에서:
h_t = tanh(W_hh * h_(t-1) + W_xh * x_t + b)

- h_t: 현재 은닉상태(현재의"기억")
- h_(t-1): 이전 은닉 상태 (과거의 "기억")
- x_t: 현재 입력
- W_hh, W_xh: 학습되는 가중치

> 딥러닝과 선형대수학에서 가중치 행렬(W)의 이름에 붙는 첨자(Subscript)는 보통 **'어디서(From) 와서 어디로(To) 가는가'**를 나타내는 이정표 역할을 한다. 

> $W_{xh}$ (x to h):출발: 현재 시점의 입력값 x_t 도착: 현재 시점의 은닉 상태 
> h_t의미: "현재 들어온 데이터(x)를 새로운 기억(h)에 얼마나, 어떤 방식으로 반영할 것인가?"를 결정하는 가중치입니다.
>즉, **현재 입력 x_t를 hidden state 공간으로 변환하는 가중치로서 지금 들어온 토큰 정보를 얼마나 반영할지! 를 정하는 가중치이다**
> 
> 
> $W_{hh}$ (h to h):출발: 이전 시점의 은닉 상태 h_{t-1}  
> 도착: 현재 시점의 은닉 상태 h_t의미: "과거의 기억(h)을 현재의 기억(h)으로 얼마나 이어받을 것인가?"를 결정하는 가중치입니다.
> 즉, **이전 hidden state를 현재 hidden state 계산에 반영하는 가중치로 이전까지 기억하는 정보를 어떻게 이어 받을지 를 결정하는 가중치이다**


hidden state의 의미: '지금까지 본 정보를 요약하고 있는 기억! (내부 메모리) 

  h_t = f(W_x x_t + W_h h_{t-1} + b)                                                                                                                                                               
                                                                                                                                                                                                   
  여기서 의미는 이렇습니다.                                                                                                                                                                        

  - x_t: 현재 들어온 토큰 정보                                                                                                                                                                     
  - h_{t-1}: 바로 직전까지의 문맥 요약                                                                                                                                                             
  - h_t: 현재 토큰까지 반영한 새로운 문맥 요약


 h_t = tanh(W_x x_t + W_h h_{t-1} + b) 


 RNN의 기본 hidden state는 위와 같이 계산한다. 하지만, h_t (현재 시점의 은닉 상태)를 만드는 함수와 h_t로 부터 최종 출력을 만드는 함수는 다를 수 있다

 tanh는 hidden state 자체를 만드는 활성화함수라면, 이렇게 구한 h_t에 대해 출력이 필요하면 별도의 출력층을 만든다!

y_t = g(W_y h_t + c) 

이때, 함수 g는 문제에 따라 달라진다(다중: 소프트맥스, 이항: 시그모이드, 회귀: linear(활성화함수 없이)..)

  - h_t를 만드는 데는 보통 tanh                                                                                                                                                                    
  - h_t를 바탕으로 출력 y_t를 만들 때는 문제에 맞는 함수 사용



W_hh 가 고정되었다고 해서, 즉, 가중치가 같다고 해서 이전 기억이 항상 같은 정도로 반영되는 것은 아니다.
하지만, 이러한 문제가 있다.

- 어떤 정보는 오래 기억해야 함                                                        
- 어떤 정보는 빨리 잊어야 함                                                          
- 그런데 vanilla RNN은 이를 유연하게 조절하기 어려움                                  
- 그 결과 장기 의존성 문제, gradient vanishing/exploding 문제가 생김  

기본 RNN에서는 정말로 “이전 정보를 얼마나 남길지”를 상황별로 정교하게 조절하는 능력이 부족하다. 왜냐하면 W_hh는 고정이고, 별도의 게이트가 없기 때문이다.

 이 한계를 보완하려고 나온 게 LSTM과 GRU예요.                                          
  이 모델들은 forget gate, input gate, update gate 같은 장치를 둬서, 문맥에 따라 이전 정보를 얼마나 유지할지 더 동적으로 조절한다. 


    - 기본 RNN: 같은 규칙으로 계속 업데이트                                               
  - LSTM/GRU: 문맥에 따라 기억/망각 정도를 더 유연하게 조절


정리:   W_hh는 모든 시점에서 공유되는 고정 파라미터이지만, 실제 반영 정도는 매 시점의 입력 x_t  와 이전 상태 h_{t-1} 가 달라지므로 항상 같지 않다. 다만 기본 RNN은 이를 문맥별로 정교 하게 조절하는 능력이 부족하다는 한계가 있다.

왜 초반 정보가 사라지기 쉬운가                                                        
  W_hh가 고정이라는 말 자체보다 더 중요한 건, 같은 변환이 계속 반복 적용된다는 점입니다.  반복 적용되면 어떤 성분은 점점 줄어들고, 어떤 성분은 과도하게 커질 수 있다.        

∂L/∂h_k = ∂L/∂h_t · ∂h_t/∂h_{t-1} · ∂h_{t-1}/∂h_{t-2} · ... · ∂h_{k+1}/∂h_k           
                                                                                        
  여기서 각 항에는 대체로 W_hh와 tanh'가 들어간다.                                    
  그래서 시간이 길어질수록 이 값들이 계속 곱해져서:                                     
                                                                                        
  - 1보다 작은 방향이면 gradient가 점점 0에 가까워짐 vanishing gradient                 
  - 1보다 큰 방향이면 gradient가 커짐 exploding gradient                                
                                                                                W_hh가 고정이다보니, 반복해서 곱해지면 어떤 성분은 점점 줄어들고 어떤 성분은 과도하게 커질 수 있다. 
                                                                                      
  이게 장기 의존성 문제의 핵심 수학적 배경이다.                    

## RNN의 한계

기울기 소실 문제: 역전파 학습 시 Grad가 시간에 따라 기하급수적으로 감소하는 현상 -> grad가 작아진다? -> 즉 w 학습이 잘 안된다.


> RNN은 똑같은 가중치 W_hh를 매 단계마다 반복해서 곱합니다. 수식을 아주 단순화해서 보면, n번째 시점의 기억은 대략 다음과 같은 형태가 됩니다.
> 
> $$h_n \approx (W_{hh})^n \times h_0$$
> 
> 
> 여기서 문제가 발생합니다.W_hh의 값이 1보다 작을 때 (예: 0.9): 0.9를 100번 곱하면 거의 0에 수렴합니다. ($0.9^{100} \approx 0.000026$)
> 
> 활성화 함수(tanh)의 영향: RNN은 보통 tanh를 쓰는데, 이 함수는 미분값이 최대 1입니다. 층을 거듭할수록 미분값이 계속 곱해지면서 0에 가까워집니다.
> 
> 결과: 학습을 위해 뒤에서부터 앞으로 미분값(오차 정보)을 전달할 때, 앞쪽으로 갈수록 정보가 사라져 버립니다. 이를 기울기 소실이라고 합니다.


장기 의존성 문제(결국 기울기 소실이 장기 의존성 문제로 연결된다): 시퀀스가 길어질 수록 초기 정보가 후반부에 미치는 영향력이 사라진다. 왜? 기울기 소실 문제로 인
해 오랜 시점과 관련된 파라미터들은 잘 학습되지 않는다. 

이 문제를 해결하기 위해 Deep RNN, BiRNN -> 하지만 잘 해결되지 않았다. 

> Deep RNN: 기본 RNN을 수식으로 쌓아 더 학습 많이하게 하여 표현력을 높임  
> BiRNN: 시퀀스를 양방향으로 처리하여 과거와 미래 정보 모두 해결해여 초반의 정보와 관련된 w가 잘 업데이트 되지 않는 기울기 소실 문제를 해결하고자 했음

-> 성능은 그다지...

그래서 나온게 LSTM & GRU이다



## RNN 활용 task

1. one  to many (하나의 토큰을 여러개의 토큰으로) : 이미지 캡셔닝
> 하나의 입력 이미지를 받아 그 이미지를 설명하는 자연어를 만드는 task이다. 

물론 전통적인 이미지 캡셔닝은 CNN이 이미지를 보고 특징 추출 RNN이 그 특징을 바탕으로 문장 단어 생성하고! 이런 방식으로 썼다


2. many to one: 감정 분류, 텍스트 분류

3. many to many (동일한 토큰 길이): 입력 n개 토큰이면 출력도 n개의 토큰
> 품사 태킹

4. many to many (다른 길이): 입력이 n개의 토큰이면 출력은 m개의 토큰
> 기계번역 (seq2seq)

## RNN 발전 흐름

RNN → Deep RNN → BiRNN → LSTM → GRU


이렇게 발전은 성능 해결하려고 하는 것이고
그 핵심은 긴문장을 못한다!! 왜? -> 기울기 소실 때문에

기울기 소실일어나는 이유: 연쇄미분의 특성상 초기 w가 업데이트가 안됌


이러한 문제점을 attention -> transformer가 해결했다.
