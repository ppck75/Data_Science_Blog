앙상블 모형에는 Bagging 알고리즘과 Boosting 알고리즘이 있다. 

---

## Bagging이란 무엇인가?

Bagging(Bootstrap Aggregating) 알고리즘이란 
Bootstrap 샘플링을 사용하여 여러 개의 모델을 학습하고
그 각각의 결과를 결합하여 최종 예측을 수행하는 알고리즘이다. 

즉, 원천 데이터 -> 여러개의 부스트랩 샘플 -> 각각 여러개의 모델 -> 모델을 aggregate하여 모델 앙상블! (분류는 투표 방식으로 예측은 모델의 평균 방식으로)

    배깅 알고리즘을 통해 만들어진 대표적인 모형이 '랜덤포레스트 모델'이다. 

### Bootstrap은 무엇인가?

부스트랩은 리샘플링 기법으로 추가적인 외부 데이터 없이 (내가 현재 가진) 원본 데이터셋을 사용하여 새로운 데이터셋을 생성하는 기법이다.

데이터가 없는 상황, 외부 데이터를 가져와서 데이터를 늘리기 어려운 상황에서 원본 데이터를 가지고 새로운 데이터를 만드는 것을 부트스트랩이라고 한다. 

    즉, 원본을 그대로 사용하는 것보다는 샘플을 가상화해서 늘리는게 그렇지 않은 것보다는 성능이 좋아지는 경우가 많다. 이 아이디어를 DT에 적용한 것이 배깅이다. 

---

## Boosting 이란 무엇인가?

부스팅(Boosting) 알고리즘이란 약한 학습기를 순차적으로 학습 시키면서 각 학습기가 이전 학습기의 성능을 향상시키는 방식으로 앙상블한 모형이다.   
*배깅을 통한 알고리즘보다 일반적으로 성능이 좋다.*

    이전 분류기가 잘맞추는 데이터는 그냥 두고, 못 맞추는 것을 그 다음 단계의 분류기가 학습하고... 이걸 반복한다. (직전 약한 분류기가 분류하지 못한 데이터에 집중한다)

### Gradient Boosting 이란 무엇인가?   

직전 분류기가 분류하지 못한 데이터에 집중하여, 다음 분류기는 그 데이터를 올바르게 분류하는 데에 주력하는 모델이다. 
> 직전 분류기의 '잔차'를 학습한다.  (즉, 직전 분류기에서 분류하지 못한 데이터를 잔차로 정의하고 이를 학습한다. )

> Gradient Boosting은 계속해서 이전 모델의 약점을 보완해간다.

### Boosting과 Gradient Boosting의 차이 

**“Boosting은 큰 범주, Gradient Boosting은 그 안의 한 방식”이다.**

🔹 Boosting의 핵심 아이디어

“약한 학습기(weak learner)를 순차적으로 학습시켜
이전 모델이 틀린 부분에 더 집중하도록 만들자”

공통 특징:

- 모델을 순차적으로 학습

- 이전 모델의 실수에 가중치를 둠

- 결과를 가중합하여 최종 예측

👉 이게 바로 Boosting 패러다임

| 알고리즘              | 핵심 아이디어                   |
| ----------------- | ------------------------- |
| AdaBoost          | 오답 샘플에 가중치 증가             |
| **Gradient Boosting** | 손실함수의 기울기(gradient)를 학습   |
| **XGBoost**           | Gradient Boosting + 최적화   |
| LightGBM          | Gradient Boosting + 효율성   |
| CatBoost          | Gradient Boosting + 범주 처리 |

🔹 AdaBoost (초기 Boosting)
작동 방식

1. 모든 데이터 동일 가중치

2. 모델 학습

3. 틀린 샘플 가중치 ↑

4. 다음 모델은 어려운 샘플에 집중


👉 직관적이지만 유연성 낮음


🔹 Gradient Boosting (일반화된 Boosting)
핵심 전환점 ⭐

“왜 굳이 샘플 가중치를 조정해야 하지?
→ 그냥 손실함수의 기울기를 직접 학습하자.”


Gradient Boosting은:

1. 현재 모델의 예측 오차 계산

2. 그 오차의 기울기(gradient) 계산

3. 다음 트리는
👉 이 기울기를 예측하도록 학습

*요즘 실무 논문에서 말하는 Boosting 모델은 거의 항상 Gradient Boosting이다*

정리하자면,  
Boosting은 약한 학습기를 순차적으로 결합하여 성능을 향상시키는 앙상블 학습 전략을 의미하며, Gradient Boosting은 손실함수의 기울기를 기반으로 이전 모델의 오차를 보정하는 방식으로 Boosting을 일반화한 알고리즘이다
