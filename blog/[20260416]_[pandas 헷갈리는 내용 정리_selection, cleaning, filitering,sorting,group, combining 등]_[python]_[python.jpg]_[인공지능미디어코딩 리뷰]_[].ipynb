## 데이터 프레임의 특정 부분 선택하기 **selcetion**


데이터프레임에서 원하는 행과 열을 골라내는 방법

| 방법 | 기준 | 예시 |
|------|------|------|
| `df['열이름']` | 열 이름 | `mydata['사고건수']` |
| `df[['열1','열2']]` | 여러 열 이름 | `mydata[['사고건수','사상자수']]` |   -> 데이터프레임으로 나오겠지
| `df.iloc[행, 열]` | **위치(숫자)** 기준 | `mydata.iloc[0, 2]` → 1행 3열 |
| `df.loc[행, 열]` | **이름(레이블)** 기준 | `mydata.loc[0, '사고건수']` |

loc[0, '열이름']: "이름표가 **0**이라고 써진 행에서 '열이름' 칸을 보여줘." (이름 기준)

iloc[0, 2]: "데이터의 맨 첫 번째(0번째) 행에서 세 번째(2번째) 열을 보여줘." (순서 기준)

> 근데 loc 할 때, 행은 보통 숫자로된 인덱스인 경우가 대부분이고, 열은 보통 글자로된 변수명을 갖기 때문에 loc(0,'사고건수') 이런 식의 경우가 많다!

````python
# '이름' 컬럼을 행의 이름(인덱스)으로 설정했다고 가정
# 인덱스: ['홍길동', '김철수', '이영희']
# 컬럼: ['나이', '사고건수']

# 1. loc 사용 (이름 기준)
mydata.loc['홍길동', '사고건수']  # 가능! (이름표를 보고 찾아감)
mydata.loc[0, '사고건수']         # 에러! (0이라는 이름표를 가진 행이 없음)

# 2. iloc 사용 (순서 기준)
mydata.iloc[0, 1]                # 가능! (이름이 뭐든 첫 번째 줄의 두 번째 칸)
````

### loc와 iloc를 쓰는 이유?




df['열'], df[['열1','열2']]와 결정적으로 다른 차이는
**행과 열을 동시에 잡을 수 있는가?** 이다.

df['열이름']은 오직 열만 선택할 수 있다. 행을 선택하려면 df['열이름'][0] 처럼 두 번 작업해야 한다

loc와 iloc는 콤마 하나로 어떤 행의 어떤 열을 가져올 지 결정할 수 있다.


**또한, 슬라이싱이 강력하다.**  


df['열이름']은 여러 열을 가져올 때 df[['열1', '열2']]처럼 일일이 이름을 다 써줘야 합니다. 하지만 loc은 **"여기부터 저기까지"**가 가능하다.

df.loc[0:5, '사고건수':'사상자수']: 0~5번 행까지, 그리고 '사고건수'부터 '사상자수' 사이에 있는 모든 열을 다 가져와!

이건 일반 인덱싱으로는 불가능한 기능이다. 


**또한, 조건문에서 많이 활용! (특정 조건을 만족하는 행의 특정 열만 가져와라 이런 식으로 )**

> df.loc[df['나이'] > 20, '이름']
>
> 만약, 인덱싱으로만 쓴다면 df[df['나이'] > 20]['이름'] 이런식으로 사용해야 한다. 

````python
# 열(변수) 선택하기
# ──────────────────────────────────────────────

# 단일 변수 선택 → Series 반환
print("=== 단일 변수 선택 (Series) ===")
display(mydata['사상자수'].head(3))

# 여러 변수 선택 → DataFrame 반환 (대괄호 2중!)
print("\n=== 여러 변수 선택 (DataFrame) ===")
display(mydata[['사고건수', '사상자수', '사망자수']].head(3))

# filter 함수로 선택
print("\n=== filter 함수로 선택 ===")
display(mydata.filter(items=['사고건수', '사상자수']).head(3))
````

````python
# 행/열 선택: iloc (위치 기반) vs loc (이름 기반)
# ──────────────────────────────────────────────

# iloc: 정수 인덱스(위치)로 선택 — i는 integer의 i
print("=== iloc: 위치 기반 선택 ===")
print("첫 번째 행, 모든 열:")
display(mydata.iloc[0, :])     # 0번 행, 모든(:) 열

print("\n처음 3행, 처음 4열:")
display(mydata.iloc[:3, :4])   # 0~2행, 0~3열 (슬라이싱)

# loc: 레이블(이름)로 선택
print("\n=== loc: 이름 기반 선택 ===")
print("0번 인덱스 행의 '사고건수':")
print(mydata.loc[0, '사고건수'])
````

## 조건으로 데이터 선택하기 **filitering**


특정 조건을 만족하는 행만 골라내는 것을 필터링이라고 한다!!

### 비교 연산자

| 연산자 | 의미 | 예시 |
|--------|------|------|
| `==` | 같다 | `mydata['사고유형구분'] == '자전거'` |
| `!=` | 다르다 | `mydata['사고유형구분'] != '자전거'` |
| `>`, `>=` | 크다, 이상 | `mydata['사고건수'] >= 5` |
| `<`, `<=` | 작다, 이하 | `mydata['사고건수'] < 3` |
| `&` | 그리고 (AND) | 조건1 **&** 조건2 |
| `\|` | 또는 (OR) | 조건1 **\|** 조건2 |

비교 연산자는 두 값을 비교하여 true/false 결과를 반환(==, !=, >, <, >=, <=)하는 논리 연산이다!

````python
mydata['사고건수'] >= 5 # 여기까지만 하면 True or False가 나오겠지
# >=는 그 값을 T 또는 F로 반환하는 비교연산자니까!
# 근데 필터링을 하려면 그 조건을 []로 한번 감싸줘야 한다!
````

````python
# 조건부 필터링
# ──────────────────────────────────────────────

# 1) 단일 조건: 사고건수 5건 이상
filtered = mydata[mydata['사고건수'] >= 5] # 5건 이상인것 찾아서 그것만 가져와서 새로운 변수로 만들어 (True인 것만 찾아서!!) 밖에 있는 []는 위치 찾기 꺽쇄이다. 
print(f"사고건수 5건 이상: {len(filtered)}행")

# 2) AND 조건: 사고건수 5~10건
filtered_and = mydata[(mydata['사고건수'] >= 5) & (mydata['사고건수'] <= 10)]
print(f"사고건수 5~10건: {len(filtered_and)}행")

# 3) OR 조건: 사고건수 10건 이상 또는 사망자수 1명 이상
filtered_or = mydata[(mydata['사고건수'] >= 10) | (mydata['사망자수'] >= 1)]
print(f"사고건수 10건+ 또는 사망자 1명+: {len(filtered_or)}행")

# 4) 문자열 조건: 특정 사고유형만 선택
filtered_type = mydata[mydata['사고유형구분'] == '자전거']
print(f"자전거 사고만: {len(filtered_type)}행")

# 5) isin(): 여러 값 중 하나에 해당하는 행 선택
filtered_multi = mydata[mydata['사고유형구분'].isin(['자전거', '보행어린이'])] # 사고유형 구분에서 자전거 보행어린이만 찾아서 다 찾아와!
# 즉, isin()이 의문문인 것이다! 물어보면, T F 돌려주는..
# 인자로 하나만 주고 싶어도 []에 넣어줘야 한다. -> 값이 하나뿐이라도 리스트로!

print(f"자전거+보행어린이: {len(filtered_multi)}행")
````

````python
# query() 메서드: 더 읽기 쉬운 필터링
# ──────────────────────────────────────────────
# 쿼리가 생겨서 쉬워짐. 

# query()는 조건을 문자열로 작성 — SQL 쿼리와 유사하여 직관적
filtered_q1 = mydata.query('사고건수 >= 5')
filtered_q2 = mydata.query('사고건수 >= 5 and 사고건수 <= 10')
filtered_q3 = mydata.query('사고유형구분 == "자전거"')
filtered_q4 = mydata.query('사고유형구분 in ["자전거", "보행어린이"]') # 컬럼 in [값들]

print(f"query 결과 - 자전거+보행어린이: {len(filtered_q4)}행")
display(filtered_q4.head(3))

# 항상 알아야 할 것은, 이 함수들은 데이터프레임 자료형에서만 쓰는 메서드이다!
# 
````

## 데이터 정렬하기 **sorting**

특정 변수 기준으로 오름 / 내림차순 정리!

````python
# 변수 기준 정렬
# ──────────────────────────────────────────────
# 정렬하는 함수가 sort()가 아니다 . sort_values()이다. 

# 사고건수 기준 오름차순 (적은 것부터)
sorted_asc = mydata.sort_values('사고건수')
print("=== 사고건수 오름차순 (상위 3행) ===")
display(sorted_asc.head(3))

# 사고건수 기준 내림차순 (많은 것부터)
sorted_desc = mydata.sort_values('사고건수', ascending=False)
print("\n=== 사고건수 내림차순 (상위 3행) ===")
display(sorted_desc[['사고지역위치명', '사고건수', '사상자수', '사고유형구분']].head(3))

# nlargest / nsmallest: 상위/하위 N개만 빠르게 추출
print("\n=== 사고건수 상위 5개 지역 ===")
display(mydata.nlargest(5, '사고건수'))


display(mydata.nlargest(5, '사고건수')[['사고지역위치명', '사고건수', '사상자수', '사고유형구분']]) # 사고건수 상위 5개 지역에서 [] 안의 칼럼값만!
````

### sort_values()  VS  sort()
sort_values()
> 값(value)을 기준으로 정렬하는 pandas 메서드입니다.
> 인자에서 준 값을 기준으로 정렬한다. 

sort()
> 파이썬 리스트의 정렬 메서드이다. 


즉, 두 함수는 쓰는 자료형이 다르다!
  - 리스트 -> sort()
  - 판다스 데이터프레임/시리즈 -> sort_values()

**정리**

  - 리스트는 sort()
  - 판다스는 sort_values()
  - 판다스에서 인덱스 정렬은 sort_index()

참고로 reset_index()는 현재 인덱스를 칼럼으로 옮기고 0부터 새 인덱스를 다시 매기는 방법이다. 

## groupby() 함수 (그룹별로 묶어서 계산)

````python
print(mydata.groupby('사고유형구분')['사고건수'].sum())
# 이건 한번에! 


# 기본 그룹화: 사고유형별 통계
# ──────────────────────────────────────────────

by_type = mydata.groupby('사고유형구분')

# 사고유형별 주요 통계
print("=== 사고유형별 사고건수 합계 ===")
display(by_type['사고건수'].sum())

print("\n=== 사고유형별 사고건수 평균 ===")
display(by_type['사고건수'].mean().round(2))

print("\n=== 사고유형별 빈도(데이터 행 수) ===")
display(by_type['사고건수'].count())
````

````python
# 다중 그룹화 + 여러 집계함수 동시 적용
# ──────────────────────────────────────────────


# 사고유형 × 사고연도별 사고건수 합계와 평균
result = mydata.groupby(['사고유형구분', '사고연도']).agg(
    사고건수_합계=('사고건수', 'sum'),
    사고건수_평균=('사고건수', 'mean'),
    사상자수_합계=('사상자수', 'sum')
).round(2)
# 물론 여기 사고연도 하나만 넣고 싶으면 리스트로 안넣어도 된다. 


print("=== 사고유형 × 연도별 통계 (상위 10행) ===")
display(result.head(10))
````

groupby()와 agg()

그룹별 요약 통계표를 한번에 여러개 만들 수 있다!!
> 여러 칼럼(하나 아님)에 대해서 서로다른 계산이 가능해진다
>
물론, 위의 사례처럼 이름 칼럼명 새로 지정하지 않고, 이런 식도 가능하다
> mydata.groupby('사고유형구분')['사고건수'].agg(['sum', 'mean', 'max'])

### 피벗테이블 (groupby의 확장 버전이다)

```pivot_table(    index="행이 될 변수",    columns="열이 될 변수",    values="계산할 변수",    aggfunc="집계함수")```

````python
# 기본 피벗 테이블
# ──────────────────────────────────────────────


# 사고유형(행) × 사고연도(열)별 사고건수 합계
pivot1 = mydata.pivot_table(
    index="사고유형구분",       # 행: 사고유형
    columns="사고연도",         # 열: 사고연도
    values="사고건수",          # 계산 대상: 사고건수
    aggfunc="sum",             # 집계 함수: 합계
    fill_value=0               # 결측치를 0으로
)
print("=== 사고유형 × 연도별 사고건수 합계 ===")
display(pivot1)
````

````python
# 고급 피벗 테이블: 여러 집계 함수 + 합계 행/열
# ──────────────────────────────────────────────

# 여러 집계 함수 적용
pivot2 = mydata.pivot_table(
    index="사고유형구분",
    values=["사고건수", "사상자수"],
    aggfunc={"사고건수": ["sum", "mean"], "사상자수": ["sum", "max"]},
    fill_value=0
).round(2)
print("=== 여러 집계 함수 적용 ===")
display(pivot2)


pivot3 = mydata.pivot_table(
    index="사고유형구분",
    values=["사고건수", "사상자수"],
    aggfunc={"사고건수": "mean", "사상자수": "sum"},
    fill_value=0
).round(2)
print("=== 여러 집계 함수 적용 ===")
display(pivot3)
````

### apply() 함수

`apply()`: **각 행이나 열에 사용자 정의 함수를 일괄 적용**  

즉, apply() 함수는 데이터프레임에서 선택한 행이나 열에 같은 함수를 반복 적용하는 메서드이다. 

apply()의 핵심 역할은:

  - 여러 열에 대해
  - 같은 함수를
  - 한 번에 반복 적용하게 해주는 것

> apply()는 함수 반복 전용 도구이다!!

참고로 apply()는 판다스 객체에 붙어 있는 메서드임. 다른데에서 못함

````python
# 1) 단일 변수에 함수 적용: 비율(%) 계산
data_pct = mydata[["사고건수", "사상자수", "중상자수"]].apply(
    lambda x: x / x.sum() * 100   # 각 열의 합계 대비 비율
).round(4)
print("=== 변수별 비율(%) 상위 3행 ===")
display(data_pct.head(3))
#  x에 열이 하나씩 들어온다. 
# 즉, 각 열마다 (3개) 지정한 lambda 함수가 적용된다 


# apply 나오면! 즉, 앞에를 대상으로 뒤의 함수를 자동적용해라(lamda아니더라도 위에서 def 정의해서 한번에 적용할 수 있는거다)
# apply 자주 사용한다. 


df.apply(lambda x: x.sum(), axis=1)
# 이 경우 행별로 적용한다. 
````

````python
# def 함수 정의 후 apply
def 심각도_평가(row):
    """사고건수 기준 심각도 분류 함수"""
    if row['사고건수'] >= 7:
        return '높음'
    elif row['사고건수'] >= 4:
        return '보통'
    else:
        return '낮음'
    
mydata['심각도'] = mydata.apply(심각도_평가, axis=1)  # axis=1: 행 단위로 적용

# 방법 2: lambda로 간단하게 (2단계 분류)
mydata['심각도'] = mydata['사고건수'].apply(lambda x: '높음' if x >= 7 else '낮음')


# 이처럼 apply()에 인자로 적용할 함수가 들어간다!!
````

## combining (데이터 합치기)

| 함수 | 방향 | 기준 | 비유 |
|------|------|------|------|
| `pd.concat(axis=0)` | 위아래 (행) | 같은 변수명(변수명이 같아야지) | 테이블 아래에 붙이기(행 기준으로 합쳐) | 
| `pd.concat(axis=1)` | 좌우 (열) | 같은 인덱스(인덱스가 같아야지) | 테이블 옆에 붙이기 |
| `pd.merge()` | 좌우 (열) | **공통 변수 값**(같은 변수가 있네?) | SQL JOIN |

```
concat(axis=0):          concat(axis=1):          merge(on='키'):
┌──────┐                 ┌──────┬──────┐         ┌──────┬──────┐
│ df1  │                 │ df1  │ df3  │         │ df1  │ df4  │
├──────┤  → 행 합치기    └──────┴──────┘         └──────┴──────┘
│ df2  │                   → 열 합치기              → 키 기준 합치기
└──────┘
```

데이터를 불러와서 붙이기가 판다스로 가능하다.  
특히, merge가 많이 사용한다



concat(axis = 1)에서 인덱스가 달라도 에러가 나지는 않는다. 옆으로 붙이기는 하는데, 인덱스가 다르면 인덱스를 맞춰가며 결합하고 없는 자리에는 NaN이 들어간다. 

마찬가지로 axis = 0 일 때에도, 에러가 나는 것이 아니라 칼럼명 맞춰서 아래로 쌓고 없는 칼럼은 NaN으로 채운다.


> 여기서 차이점! pd.concat()은 합칠 데이터를 리스트로 [] 줘야 하는데, merge()는 합칠 데이터를 리스트로 담지 않는다. 왜? concat은 여러개의 객체를 한꺼번에(2개 이상도 가능) 붙이는 방식이기 때문에 '합칠 데이터프레임들의 묶음'을 받는 것이기 때문이다. 

````python
# 데이터 합치기 실습용 샘플 생성
# ──────────────────────────────────────────────

data1 = pd.DataFrame({"변수1": [1, 2, 3], "변수2": [5, 3, 1]}) # 데이터프레임 클래스를 생성하기 위한 인자들임
data2 = pd.DataFrame({"변수1": [21, 43], "변수2": [54, 12]})
data3 = pd.DataFrame({"변수3": [224, 453, 648]})
data4 = pd.DataFrame({"변수1": [1, 2, 5], "변수3": [224, 453, 648]})

# 1) 행 방향 합치기 (위아래로)
combined_rows = pd.concat([data1, data2], axis=0, ignore_index=True)
print("=== 행 방향 합치기 (concat axis=0) ===")
display(combined_rows)

# 2) 열 방향 합치기 (좌우로)
combined_cols = pd.concat([data1, data3], axis=1)
print("\n=== 열 방향 합치기 (concat axis=1) ===")
display(combined_cols)
````

````python
# 인덱스가 맞지 않는 경우

pd.concat([data1, data2], axis=1) # 에러는 안나지만 NaN 값이 들어온다!
````

````python
# merge: 공통 변수(키) 기준으로 합치기 — SQL JOIN과 동일
# ──────────────────────────────────────────────

# inner join: 양쪽 모두에 있는 키 값만 (교집합) -> 공통적인 것만
merged_inner = pd.merge(data1, data4, on="변수1", how="inner")
print("\n=== Inner Join (교집합) ===")
display(merged_inner)

# outer join: 양쪽 어디에든 있는 키 값 모두 (합집합) -> 한번이라도 있으면 잡아
merged_outer = pd.merge(data1, data4, on="변수1", how="outer")
print("\n=== Outer Join (합집합) ===")
display(merged_outer)

# left join: 왼쪽 데이터의 키 값 기준 -> 왼쪽 기준으로!(right도 가능함)
merged_left = pd.merge(data1, data4, on="변수1", how="left")
print("\n=== Left Join (왼쪽 기준) ===")
display(merged_left)
````

### crosstab VS pivot

  - crosstab()은 범주형 데이터의 빈도표/교차표를 빠르게 만드는 데 강하다. 
  - pivot()은 데이터의 모양을 재배치하는 데 쓴다.

  즉 crosstab()은 "몇 개냐, 비율이 얼마냐" 쪽이고,
  pivot()은 "행과 열 구조를 바꿔서 표를 다시 만들기" 쪽이다


pivot은 갯수를 세는게 아니라 원래 있던 점수 값을 행 열 구조로 재배치한 것이다. 반면, crosstab은 범주 조합별 빈도, 비율, 집계 등을 보려고 한다.


참고로  **집계함수를 쓰는 건 pivot_table()**

pivot()은 모양만 바꾸고, pivot_table()은 모양도 바꾸면서 집계도 한다.
