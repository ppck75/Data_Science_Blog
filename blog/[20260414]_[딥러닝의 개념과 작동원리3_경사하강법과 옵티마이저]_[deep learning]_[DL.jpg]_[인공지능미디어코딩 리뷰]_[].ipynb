## 경사하강법

손실함수의 값을 최소화하기 위해 가중치를 조금씩 조정하는 방법이다

**수식**

W(new) = W(old) - learning_rate * dL/dW

파라미터가 100만개라면, 100만개의 축을 가진 하나의 거대한 손실함수 산맥이 형성될 것이다. 이 거대한 산맥의 모든 방향의 경사를 한꺼번에 고려해서 가장 가파른 아래 쪽으로 한 발자국 내딛는 것이다. 

1. 모든 w에 대한 미분값을 구한다 (모든 방향의 경사를 잰다)  
2. 그 경사들을 모아 하나의 벡터를 만든다.  
3. 모든 가중치를 동시에 이동시킨다

물론, 가중치는 각자의 길을 간다. 어떤 것은 양수 어떤 것은 음수로.. 하지만, 이것드링 동시에 움직인다는 것이다. 


**그런데, 경사하강법은 로컬 미니멈에 빠질 우려가 있다** (우리가 원하는 것은 글로벌 미니엄에 가는 것인데!)

3. 통합

모멘텀

모멘텀은 어느쪽으로 갈것인가를 결정할 떄 과거의 그라디언트의 방향 즉, 관성을 반영하는 식이다. 

모멘텀은 방향의 일관성을 유지하게 해주는 옵티마이저이다. 학습 과정의 진동을 감소시키고 최저점까지 효율적으로 도달 할 수 있게 한다. 


## 파라미터와 하이퍼파라미터

파라미터: 모델 안에서 계산되게 만든 변수들이다. 주로 w이다

하이퍼파라미터: (모델 안에서 업데이트 되는 것이 아닌) 모델 돌리기 위해서 우리가 사전에 지정하는 것이다

## 딥러닝 전체 흐름 

1. 모델 초기화 (Initialization) (w와 b를 무작위로 설정함)

2. 순전파 (Forward Pass) (입력 데이터를 넣고 행렬 연산을 거쳐 예측값을 만들어낸다.)

3. 손실 계산 (Loss Calculation)

4. 역전파 (Backpropagation) (기울기를 계산하여 각 가중치를 얼만큼 수정해야 하는지 파악한다)

5. 가중치 업데이트 (Optimizer) (옵티마이저를 이용하여 오차가 줄어는 방향으로 가중치 수정)

6. 반복 (Iteration/Epoch)

## 글 정리 및 최종 결론

이 글에서는 역전파로 계산한 기울기를 실제 가중치 수정으로 연결하는 경사하강법과 옵티마이저의 역할을 살펴보았다. 기본 경사하강법이 현재 위치의 기울기를 따라 손실을 줄이는 원리, 모멘텀과 적응적 학습률이 이동 방향과 보폭을 개선하는 방식, Adam이 두 접근을 결합하는 이유, 그리고 파라미터와 하이퍼파라미터의 차이를 딥러닝 전체 학습 흐름과 연결했다.

### 경사하강법의 핵심 원리

경사하강법은 손실함수의 값을 줄이기 위해 가중치를 조금씩 조정하는 방법이다. 기본 갱신식은 `W_new = W_old − learning_rate × ∂L/∂W`이며, 벡터 표기로는 `W_new = W_old − η∇_W L`이라고 쓸 수 있다. 기울기 ∇_W L은 현재 위치에서 각 가중치를 어느 방향으로 변화시킬 때 손실이 가장 빠르게 증가하는지를 나타내므로, 모델은 그 반대 방향으로 이동하여 손실을 낮춘다.

가중치가 100만 개라면 손실함수는 100만 개의 파라미터 축을 가진 고차원 공간에 정의된다. 학습 과정에서는 모든 가중치에 대한 편미분을 구해 하나의 기울기 벡터로 묶고, 각 가중치를 동시에 갱신한다. 개별 가중치는 기울기의 부호와 크기에 따라 서로 다른 방향과 거리로 이동하지만 하나의 업데이트 단계에서 함께 바뀐다.

학습률은 한 번에 이동하는 크기를 결정한다. 경사 방향이 올바르더라도 학습률이 지나치게 크면 최저점을 지나치며 진동할 수 있고, 지나치게 작으면 손실이 줄어드는 속도가 매우 느려질 수 있다. 따라서 경사뿐 아니라 이동 경로와 이동 크기를 함께 조절하는 것이 중요하다.

### 옵티마이저가 경사하강법을 개선하는 방식

기본 경사하강법은 현재 단계의 기울기만 보고 이동하므로 굴곡이 큰 손실 공간에서 지그재그로 진동하거나 수렴이 느려질 수 있다. 또한 원하는 전역 최저점이 아니라 국소 최저점 주변에 머물 가능성도 있다. 옵티마이저는 이러한 문제를 완화하면서 손실을 더 효율적으로 줄이도록 가중치 갱신 규칙을 개선한다.

| 개선 방향 | 핵심 아이디어 | 글에서 다룬 의미 |
| --- | --- | --- |
| 기본 경사하강법 | 현재 기울기에 따라 이동한다. | 단순하지만 진동하거나 수렴이 느릴 수 있다. |
| 모멘텀 | 과거 기울기의 이동 방향을 누적한다. | 일관된 방향에는 속도를 붙이고 불필요한 진동을 줄인다. |
| 적응적 학습률 | 파라미터와 상황에 따라 이동 크기를 조절한다. | 모든 가중치에 같은 보폭을 적용하는 비효율을 줄인다. |
| Adam | 모멘텀과 적응적 학습률을 결합한다. | 방향의 관성과 파라미터별 보폭 조절을 함께 사용한다. |

모멘텀은 과거 기울기의 방향을 반영하여 현재 이동 방향을 결정한다. 같은 방향의 기울기가 반복되면 그 방향으로 속도가 붙고, 서로 반대되는 기울기가 반복되면 진동이 완화된다. 적응적 학습률 계열은 모든 가중치에 동일한 학습률을 적용하지 않고 각 파라미터의 상황에 따라 갱신 크기를 다르게 조절한다. Adam은 이 두 발상을 결합한 대표적인 옵티마이저다.

옵티마이저가 더 정교하다고 해서 항상 전역 최저점을 보장하는 것은 아니다. 옵티마이저의 역할은 손실 공간에서 더 안정적이고 효율적인 경로를 찾도록 갱신 방식을 개선하는 데 있다.

### 파라미터와 하이퍼파라미터

파라미터는 모델이 데이터로부터 직접 학습하는 값이다. 신경망의 가중치 W와 편향 B가 대표적이며, 역전파와 옵티마이저를 통해 반복적으로 갱신된다.

하이퍼파라미터는 모델이 학습 과정에서 스스로 구하는 값이 아니라 연구자가 학습 전에 정하거나 검증 과정을 통해 선택하는 설정이다. 학습률, 배치 크기, 에포크 수, 은닉층의 수와 노드 수, 사용할 옵티마이저 등이 이에 해당한다. 하이퍼파라미터는 파라미터가 어떤 방식과 속도로 학습되는지를 통제한다.

### 딥러닝 학습의 전체 흐름

딥러닝의 학습은 다음 여섯 단계가 반복되는 과정이다.

1. 모델을 초기화하여 가중치와 편향의 시작값을 정한다.
2. 순전파로 입력 데이터의 예측값을 계산한다.
3. 손실함수로 예측값과 실제값의 차이를 측정한다.
4. 역전파로 각 가중치에 대한 손실의 기울기를 계산한다.
5. 옵티마이저가 기울기와 자신의 갱신 규칙을 이용하여 가중치를 수정한다.
6. 이 과정을 이터레이션과 에포크 단위로 반복한다.

순전파와 손실 계산은 현재 모델이 얼마나 틀렸는지를 확인하고, 역전파는 어떤 가중치를 어느 방향으로 수정해야 하는지 계산하며, 옵티마이저는 계산된 기울기를 실제 이동 규칙으로 바꾼다. 이 세 단계는 분리된 기능을 수행하지만 하나의 학습 사이클 안에서 연속적으로 작동한다.

### 최종 결론

딥러닝 모델이 학습한다는 것은 손실함수의 고차원 공간에서 더 낮은 값을 향해 파라미터를 반복적으로 이동시킨다는 뜻이다. 역전파는 이동에 필요한 기울기를 계산하고, 경사하강법과 옵티마이저는 그 기울기를 이용하여 실제 이동 방향과 크기를 결정한다.

따라서 좋은 학습은 단순히 기울기를 구하는 것만으로 완성되지 않는다. 현재 기울기만 따를지, 과거 방향의 관성을 반영할지, 가중치마다 보폭을 다르게 조절할지에 따라 수렴 경로와 속도가 달라진다. 모멘텀은 방향을 안정화하고, 적응적 학습률은 보폭을 조절하며, Adam은 두 원리를 결합한다.

결국 파라미터는 모델이 데이터에서 학습하는 내용이고, 하이퍼파라미터와 옵티마이저는 그 학습이 이루어지는 방식을 결정한다. 모델 초기화에서 시작해 순전파, 손실 계산, 역전파, 옵티마이저의 가중치 갱신과 반복으로 이어지는 전체 흐름을 이해하면 딥러닝을 단순한 함수 호출이 아니라 오차를 줄이기 위한 연속적인 최적화 과정으로 파악할 수 있다.
