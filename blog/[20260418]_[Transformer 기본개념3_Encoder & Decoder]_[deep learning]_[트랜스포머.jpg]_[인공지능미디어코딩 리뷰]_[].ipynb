## 트랜스포머에서 존재하는 attention 종류

  - Q = K = V : Self-Attention                                                          
  - Q = K = V + mask : Masked Self-Attention                                            
  - Q ≠ K = V : Cross-Attention 

## 인코더와 디코더의 FFNN의 역할

어텐션 층 바로 다음에 붙어있는 **FFNN(Feed-Forward Neural Network)**의 핵심 역할

1. 개별 단어의 정보 업데이트(position wise - '위치별' 수행!)
> 어텐션에서 모은 단어 정보들을 가지고 각자의 단어를 독립적으로 계산한다.
> 어텐션에서는 정보를 섞었다면, 여기서는 각 단어마다 철저히 자기자신 계산 수행
> 왜 -> 섞인 정보들을 바탕으로 현재 위치에 있는 '나'라는 단어의 최종 의미를 확정 지어서 **자기자신만의 고유한 의미 벡터**로 강화해 나가는 것이다!! 

2. 비선형성 추가를 통한 표현력 확장  
> 어텐션은 행렬곱셈 위주이다 즉, 선형 연산이다. 현실의 복잡한 언어 데이터는 단순 곱셈만으로 표현하기 힘들기 때문에 FFNN 내부에서 ReLU와 같은 활성화 함수를 통해 비선형적인 특성을 학습할 수 있도록 한다.
>
> 예를 들어 이 단어가 이 맥락에서는 이러한 복합적 의미를 갖는다! 와 같은 비선형적인 특징을 학습하도록 한다.

3. 정보의 재구성(차원 확대와 축소 과정을 통해)
> 트랜스포머에서 FFNN은 입력차원 512 -> 중간차우너 2048 -> 출력차원 512의 구조를 가진다.
> 차원을 확대하고 축소하면서 단어가 가진 미세하고 잠재적인 특징을 분석하고 핵심 정보만 남긴다. 

### 핵심: 인코더는 입력 전체를 동시에 이해하고(encoder self attention), 디코더는 하나씩 생성하며(Masked Decoder Self-Attention) 필요할 때 인코더에게 물어본다(encoder - decoder cross attention)
