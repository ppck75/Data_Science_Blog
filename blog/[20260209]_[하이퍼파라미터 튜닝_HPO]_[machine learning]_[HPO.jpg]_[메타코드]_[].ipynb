## 하이퍼파라미터 이해하기 위한 배경(Overfitting)

의사결정 나무의 예를 들어보자.  

DT 알고리즘으로 모형을 학습할 때 '언제까지 나무를 키워야 할까?'

만약, 데이터에 노이즈가 전혀 없는 이상적인 상황이라면 DT를 가장 크고 깊게 복잡한 모형으로 만드는 것이 성능을 극대화하는 것이다. 이것은 모형을 복잡하게 만든다는 것이며, 이는 곧 모형 안에 굉장히 많은 분기가 생긴다는 것이다.

만약, 노이즈가 없다면, 이런게 가능하다.  
**하지만, 이런 경우는 실제에서 없다**

현실에 있는 데이터는 노이즈와 변동성이 가득하기 때문에 우리는 의미 있는 패턴'만' 학습해야 한다. (모델을 전체 다 학습 은 곧 노이즈까지 학습하는 것을 의미하는데, 이런 경우 모델 성능이 떨어진다)


DT의 분기가 너무 많으면, 깊으면 노이즈나 변동성까지 학습 - 과적합이다.  
    잘못된 분류는 자연적으로 발생한 노이즈로 생각하게 된다. 그런데 자연적으로 발생한 노이즈를 모형이 모두 학습한 상태가 과적합이다.  

    이는 학습하면 안되는 것까지 학습한 상태이다. 이 경우 노이즈까지 학습하면 성능이 정말 안좋아진다. 

우리가 ML로 모형을 만들 때에는 적절한 양의 성장을 하고 더이상 모형이 복잡해지지 않도록 통제하는 여러 기법들이 반드시 필요하다.(즉, 튜닝이 필요하다. DT를 예시로 들면 노드 수라던지 깊이라던지...) 

우리는 데이터의 사이즈, feature의 수에 따라 달라지기에 이 데이터에 대한 이해를 가지고 나무의 크기를 결정해야 한다.

## Hyperparameter란 무엇인가?  
**핵심**  

‘하이퍼파라미터는 모델의 학습과정을 제어하는 사용자 설정변수  
    쉽게 말하면, 학습 옵션이다.

**학습에 영향을 주는 제약조건인 하이퍼파라미터!**

우리는 하이퍼파라미터를 조절해서 나무의 크기와 모형을 조절할 수 있다.
DT의 예시를 들어보자면, 다음과 같은 옵션이 있다. 
> Criterion: 불순도 기준   
> Splitter: 각 노드에서 분할을 수행 방법  
> max_depth: 트리의 최대 깊이  
> min_samples_split: 노드의 최소 샘플 수  
> min_samples_leaf: 리프의 최소 샘플 수  
> max_features: 사용할 특성의 최대 개수  
> min_impurity_decrease: 최소 불순도 감소량  
> class_weight: 클래스 가중치  
>random_state: 난수 생성기의 시드  

우리는 이러한 하이퍼파라미터를 설정하여서 최적의 옵션을 가진 최적의 모형을 찾아가야 한다. 



---
## Hyperparameter Tunning 어떻게 하나? 

하이퍼파마리터 최적화(HPO) 또는 하이퍼파라미터 튜닝은 - > 최적의 옵션을 찾아주는 방법이다. 그럼 이걸 우리가 하나하나 찾아가면서 많은 경우의 수를 계산해야 하는가..?

> 아니다!! 최적의 하이퍼파라미터를 찾아주는 방법이 있다. 

대표적으로 **GridSearchCV**가 있다. 

# HPO 실습  

*메타코드 부트캠프 3주차에서 직접 실습한 코드 내용입니다*

```python
# DecisionTreeClassifier 뒤에 ?를 넣어주면, 파이썬은 매우 친절해서 이 알고리즘이 받을 수 있는 하이퍼파라미터의 개념들을 쫙 보여준다. 

DecisionTreeClassifier? 

# 싸이킷 런에 있는 다음과 같은 파라미터를 받을 수 있으며, 대부분이 학습에 영향을 주는 파라미터이다.  
# 또한 내부적으로 가지고 있는 특성들도 보여준다. 
```

```python
# 이것은 수기로 하이퍼파라미터를 설정해 보는 것이다. 
clf_mannual = DecisionTreeClassifier(criterion='gini',
                       max_depth=1,
                       min_samples_split=5, # sample보다 leaf가 크면 말이 안된다. 샘플(노드)로 분할하다가 더이상 분할 못하면 리프가 되는 것이기 때문이다. 
                       min_samples_leaf=3, # 당연히 leaf가 samples보다 작아야 할 것이다. 에러가 날 수도?!
                       splitter='random',
                       random_state=4) 

clf_mannual.fit(X_train, y_train)

y_pred_mannual = clf_mannual.predict(X_test)
ac_mannual = accuracy_score(y_test,y_pred_mannual)
print(ac_mannual)

# 이 경우 모델의 정확도가 하이퍼파라미터를 기본 디폴트값으로 학습했을 때보다 성능이 더 안좋아졌다.
# 모형의 정확도를 높이기 위해서는 셋팅을 바꿔야 한다(물론, 내가 판단을 해야겠지)
```

이런 방식으로 직접 하이퍼파라미터를 결정할 수 있다. 하지만, 이게 최적의 옵션인가? 그건 모른다. 그러므로 우리는 최적의 옵션을 찾아주는 방법을 사용한다. 

**가장 우수한 모형을 자동으로 찾아주는 HPO를 해보자!**

GridSearch은 모든 가능한 경우의 수를 찾는 방식이다. 

성능 측면에서는 좋은 알고리즘인데, 계산량이 많다는 한계가 있다. 단, 트리 모형에서는 걍 이걸 써도 전혀 무방하다. 

```python
# GridSearch를 HyperParameter의 범위를 한정해보자. (dict 형태로 넣을 것이다. )

param_grid = {
    "criterion" : ['gini', 'entropy'], # 이 두가지 중에 하나를 찾아줘(참고로 3가지 계산방법 있음)
    "max_depth" : [2,3,4,5], # 2,3,4,5 즉, 최대 5까지 가질 수 있는 depth로 하자. 
    "min_samples_split": [2, 5, 10], # 2,5,10 중에 찾아줘!
    "min_samples_leaf": [1, 2, 4] # 1,2,4중에 찾아줘!
}

# grid search가 가능한 모든 경우의 수를 만든다. 단, 우리가 위에서 설정한 조합으로만 그 경우의 수를 한정하는 것이다. 
# 우리가 설정하지 않은 하이퍼파라미터들은? -> 디폴트 값으로 찾겠지. 
```

*여기서 알 수 있는 점은 모든 것이 자동은 아니라는 점이다.. 우리가 하이퍼파라미터 경우의 수를 주면, 거기서 찾는 것이다.*  
> 생각해보면, 모든 옵션의 경우의 수를 다 고려하면 계산이 엄청 오래 걸릴 것이다..

```python
clf_grid = DecisionTreeClassifier( random_state= 42 ) 
# grid에 없었기 때무에 random_state= 42 로 설정해준다. 

#  중요
grid_search = GridSearchCV(clf_grid, param_grid, cv = 5) 

# 중간단계가 하나 더 필요하다. 
# cross validation은 5로 주겠다. 
# 우리가 이전에는 1단계에서 바로 학습을 시켰다. fitting을 바로 시켰는데, hpo를 해야하기 떄문에 그 범위를 넣고 search 해준다.
# 여기서 search는 최적의 옵션 조합을 찾는 search일 것이다. 
# CV = 5는 전체 데이터를 5개로 분류하고 4개를 학습 데이터로 쓴다. 이걸 5개의 모든 경우의 수에서 각각 수행한다! 이렇게 보면 된다. 
```

```python
# HyperParameter를 찾고, 이걸 가지고 fitting이 모두 수행하는 부분임!(모두 수행된다.)

grid_search.fit(X_train, y_train) 

# 이걸 하는 순간에 search가 모든 경우의 수를 계산을 하고, CV 5덩어리를 가지고 정확도를 검정해서 최적의 하이퍼파라미터를 찾고 그걸 가지고 학습을 하는 것이 한번에 이뤄지는 것이다.
# 그러니까 시간이 오래 걸리는 부분은 이 코드이겠다. 

print("Best Hyper-parameter", grid_search.best_params_)
print("Best Score", grid_search.best_score_)
```

> Best Hyper-parameter {'criterion': 'gini', 'max_depth': 3, 'min_samples_leaf': 1, 'min_samples_split': 2}
Best Score 0.9224137931034484 

```python
# HPO만들어진 모형의 정확도 계산 
best_model = grid_search.best_estimator_

y_pred_grid = best_model.predict(X_test)
accuracy_grid = accuracy_score(y_test, y_pred_grid) # 정확도 계산
print('Accuracy Grid :', accuracy_grid)

# 일반적으로 학습할 때의 데이터보다 낮아진다. 근데 이 데이터는 0.92에서 0.94로 늘긴했네.. 하지만, 실제로 데이터 수가 매우 많으면 테스트 데이터 셋에서 데이터가 소폭 낮아지는게 일반적이다. 
```

> Accuracy Grid : 0.9444444444444444

---

grid_search.best_score_는
“최적 하이퍼파라미터 조합에 대해
교차검증(cv)에서 얻은 평균 성능 점수”을 의미한다. 

> cv=5 → 5겹 교차검증이라는 의미이다. 
GridSearchCV는 내부적으로 이렇게 함 👇
하이퍼파라미터 조합 하나 선택

데이터를 5개 폴드로 나눔

각 폴드에서:  
  
4개 폴드로 학습  

1개 폴드로 검증  

Accuracy 계산  

5개의 Accuracy 평균이 

모든 조합 중 평균이 가장 높은 것 선택  

👉 그 최고 평균값이 바로:  grid_search.best_score_이다. 
즉, 최적의 의사결정나무 하이퍼파라미터 조합에서
**5겹 교차검증 기준 평균 Accuracy가 약 92.24%**였다.

하지만, 이것은 테스트 성능이 아니다. 훈련 데이터 내부의 교차검증 평균 성능이야. 
 best_score_는 모델 선택을 위한 기준값이지
최종 일반화 성능이 아니다.  


best_score_ = “CV로 추정한 성능”
test score = “진짜 처음 보는 데이터 성능” (모델 평가에서 우리가 보는 지점)  

*여기서 cv=5란*  

전체 데이터를 8:2로 나눈 뒤,
그중 80%(학습 데이터)만 가지고
다시 5개로 쪼개서 교차검증을 한다.  

전체 데이터 (100%)  

X_train (80%)   ← 여기만 학습에 사용  
X_test  (20%)   ← 완전히 잠가둔 데이터  

gridsearchCV 내부에서 일어나는 일  
X_train (80%)
 ├─ Fold 1 (train 64% / val 16%) 4개 폴드로 학습하고 하나로 검증...  
 ├─ Fold 2 (train 64% / val 16%)  
 ├─ Fold 3 (train 64% / val 16%)  
 ├─ Fold 4 (train 64% / val 16%)  
 └─ Fold 5 (train 64% / val 16%)  

이 결과로 5겹 교차검증으로 추정한 평균 성능을 내보낸다. 

---

**cv=5는 어떤 기준일까?**  

cv 값은 ‘정답’이 있는 게 아니라,
데이터 크기 · 계산비용 · 추정 안정성의 균형으로 정한다.

그래서 5와 10이 표준처럼 쓰인다.


📘 교재 기준

계산량이 적다

설명하기 쉽다

실습 환경(노트북)에서 무리 없음

대부분의 모델에서 충분히 안정적

👉 그래서 교육용 표준값 = 5


현업에서는? 

cv = 5 → 디폴트  

cv = 10 → 데이터 작거나 중요한 모델

논문에서는? 

 | 상황        | cv 선택             |
| --------- | ----------------- |
| 일반 ML 논문  | 5-fold 또는 10-fold |
| 데이터 작음    | **10-fold**       |
| 매우 작은 데이터 | LOOCV             |
| 불균형 데이터   | Stratified k-fold |
| 모델 선택 엄격  | Nested CV         |

그리고 보통 이렇게 씀:

“We used 10-fold cross-validation to evaluate model performance.”

👉 5 또는 10이면 reviewers가 전혀 문제 삼지 않음
