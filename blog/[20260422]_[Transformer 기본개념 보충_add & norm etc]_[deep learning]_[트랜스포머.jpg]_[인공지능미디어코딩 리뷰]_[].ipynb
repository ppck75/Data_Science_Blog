# Transformer 기본개념 4

## Add(잔차연결) & Norm(정규화)

*인코더와 디코더 내부 구조를 보면 잔차연결과 정규화가 반복된다*

인코더 내부 구조를 간략하게 설명해보면,

512차원의 임베딩을 8개로 나누어서(멀티헤드 어텐션을 위해) 8개 헤드에서 각각 다른 관점의 attention이 일어난다. (QKV 구하고 내적하고.. attention value가 만들어진다. 즉, 서로 다른 토큰들간의 문맥 관계를 이해한 벡터로 바뀐다. )


그 후 add norm을 거친다. 

잔차연결: 원래 입력을 다시 더해 정보 손실을 방지한다

정규화: (add에서 더했는데, 값이 너무 커질 수 있으니) 정규분포로 만들어준다!

즉, add는 원래 것도 반영해줘!(attetoin을 통해서 구한 attention value에 원래 값도 반영해줘)


norm은 add 했는데 분포가 너무 왔다갔다 커지네.. 정규분포로 만들자!


그 후 다시 FFNN 진행된다. 그리고 또 다시 잔차연결과 정규화가 진행된다. 


**이 과정 즉, 멀티헤드 셀프어텐션 -> 잔차연결 정규화 -> FFNN -> 잔차연결 정규화**가 num_layer 만큼 반복된다. 당연히 디코더도 마찬가지이다!!


num_layer가 일반 트랜스포머에서 6층이다.

1층에는 단어 수준의 관계 파악한다고 하면, 6층에서는 문장 전체의 의미와 뉘양스까지 파악하며 깊은 이해로 나아간다. 

| 단계 | 용어 | 설명 | 세부 내용 |
|:---:|:---|:---|:---|
| 1 | **Multi-Head Self-Attention** | 모든 단어가 서로를 바라보며 관계 파악 | 8개 헤드가 각각 다른 관점으로 분석 |
| 2 | **Add & Norm (잔차 연결 + 정규화)** | **Add**: 원래 입력을 다시 더해 정보 손실 방지<br>**Norm**: 숫자 스케일을 일정하게 맞춰 안정화 | 정보 손실 방지 및 학습 안정성 확보 |
| 3 | **Position-wise FFNN** | 각 단어의 표현을 개별적으로 풍부하게 | 512 → 2048 → 512 확장/축소, 비선형 패턴 학습 |
| 4 | **Add & Norm** | FFNN 결과에도 잔차 연결 + 정규화 적용 | 매 단계마다 반복하여 학습 안정성 확보 |
