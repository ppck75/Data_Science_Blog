# Attention 알고리즘에 대해 알아보자




## 정의

attention을 한 문장을 설명하면,

**Query와 Key간의 유사도를 바탕으로 Value를 구한다!! 로 정리할 수 있다**

좀 더 정확히 말하면 Q 벡터와 K 벡터로 유사도의 가중치를 계산하고, 그 가중치를 Value 벡터를 가중합하여 출력값을 만드는 연산이다!!

*self attention이든, cross attention이든 아니면 seq2seq의 attention이든, 위의 큰 틀을 모두 따른다. 즉, Q와 K를 곱하고 V를 가중합 해서 출력을 만드는 구조는 모두 똑같다. 다만, 차이를 만드는 것은 무엇이 Q, K, V이냐? -> 이게 구조마다 다른것이다!!*

  - Self-Attention                                                                      
    Q, K, V가 모두 같은 시퀀스에서 옴                                                   
    즉, 한 문장 안의 단어들이 서로를 참고                                               
  - Cross-Attention                                                                     
    Q는 디코더 쪽에서 오고                                                              
    K, V는 인코더 출력에서 옴                                                           
    즉, 디코더가 입력 문장을 참고                                                       
  - RNN + Attention                                                                     
    보통 Q는 디코더의 현재 hidden state                                                 
    K, V는 인코더의 hidden state들                                                      
    즉, 디코더가 인코더의 여러 시점 중 어디를 볼지 정함



**이렇게 모두 구조가 다르지만, 결국 Q로 질문하고 K로 비교하고, 그 중요도 만큼 V에서 정보를 가져오는 것은 다 똑같다는 것이다!*


Q≠K=V
 : Cross-Attention 
    > 디코더와 인코더 사이에서 일어나는 attention!!

 
Q=K=V
 : Self-Attention
    > 인코더에서 일어나는 어텐션! 또는 masked Decoder attention도 여기에 해당한다. 결국 한 시퀀스에서 오기 때문이다. Q = K = V 인데 attention이 볼 수 있는 위치를 가린 self attention인 것이다

## 트랜스포머에서 존재하는 attention 종류

  - Q = K = V : Self-Attention                                                          
  - Q = K = V + mask : Masked Self-Attention                                            
  - Q ≠ K = V : Cross-Attention
