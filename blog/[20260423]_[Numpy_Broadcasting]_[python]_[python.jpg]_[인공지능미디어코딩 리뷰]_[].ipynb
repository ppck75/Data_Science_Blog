# Numpy 브로드캐스팅

array에서 계산을 효율적으로 해주는 것이 브로드캐스팅이다. 

array의 shape이여도 빈 shape는 대충 잘 복사해서 계산을 해준다~~


array를 쉽게 해석해보면, 어릴 때 배운 사칙연산의 계산 원리를 생각해 보면 쉽다. 

우리가 256 + 3을 할 때 3은 맨 오른쪽으로 붙여서 2와 더한다. 마찬가지로 브로드캐스팅 할 때에도 shape의 순서를 오른쪽으로 붙인다. 
> 즉, shape(B,L,D) array와 shape(1,1) array를 계산한다고 하면 (1,1)을 오른쪽으로 밀어서 한다!!

(2,3) shape와 (3) shape array의 곱셈은 3 shape를 오른쪽으로 밀고 shape 차원이 적은 쪽을 다 1로 채워 넣는다. 즉, (2,3) 과 (1,3)을 한다. 그래서 축별로 비교를 하는데, shape가 같은 축은 잘 연산 가능하고 한 쪽이 1이면 나머지 한 쪽에 맞춰서 복사를 해준다. -> 이떄 브로드캐스팅이 일어난 것이다


근데, 만약 이렇게 오른쪽으로 다 밀고 비교할 때 두 shape가 둘 다 1도 아니고 숫자도 다르면 에러가 난다. 
즉,(2,3) array와 (2,) array는 (2,3)과 (1,2) 가 되는데 2와 3을 비교할 때 shape 다르고 둘 다 1도 아니여서 에러가 난다. 

결국 브로드캐스팅을 할 때에는 shape가 오른쪽 정렬이 되었을 때를 고려하면? 헷갈리지 않고 다차원 array의 연산을 설계할 수 있다. 

또한, 스칼라와 배열의 연산에서도 브로드캐스팅이 적용된다. (가장 단순한 브로드캐스팅)

A = np.array([[1, 2, 3],
              [4, 5, 6]])

A * 2 
> 모든 원소에 2를 곱함 # element wise하게 원소별 곱을 하는 것이다. 


 (3,1) + (1,3) → (3,3)  
col = np.array([[1], [2], [3]])   # shape (3,1)  
row = np.array([[10, 20, 30]])    # shape (1,3)  

col + row -> (3,3) shape의 array

### 브로드캐스팅 정의

서로 **shape이 다른 배열끼리 연산**할 때, NumPy가 자동으로 크기를 맞춰주는 규칙  
명시적으로 배열을 복제하지 않아도 되므로 메모리 효율적

#### 브로드캐스팅 규칙 (오른쪽 축부터 비교)

1. 두 축의 길이가 **같거나**
2. 둘 중 하나가 **1**이면 호환
