# 임베딩

컴퓨터가 다룰 수 있는 것은 오직 숫자 뿐이기 떄문에, 우리는 텍스트를 숫자로 바꿔줘야 한다.
텍스트를 숫자로 바꿔주는 방법에는 3가지가 있는데, **one-hot encoding, TF-IDF, 밀집벡터(word2vec)** 이 있다.

> 우리가 하는 작업(원핫, TF IDF, 임베딩 모두)의 궁긍적인 목표는 *문장 또는 문서*를 하나의 숫자 묶음인 벡터로 만드려고 하는 것이다.  
> 단어를 벡터화할 수도 있지만, 문장 벡터(sentence verctor)를 만들기 위한 수단!으로 이해하면 될듯  
> 단어 벡터는 재료이다. 



## one-hot encoding

해당 단어가 있으면1, 없으면 0으로 인코딩하는 방식이다.  
이렇게 하면, 벡터의 차원이 곧 어휘의 수이다. 

원핫 인코딩은 겹치는 단어가 적으면 유사도가 낮게 나온다. 

이 방식의 문제점은

> 1. 우리가 사용하는 vocab에는 단어가 매우 많다.
> 만약 vocab 사이즈가 10만이면, 한 문장(documnet)에서 몇개만 1이고 나머지는 다 0이다 -> **spares vector**의 문제
>
> 2. 비록 단어 형태는 다르지만, 같은 맥락 안에서 쓰이는 단어들 즉, 의미적 유사성을 포착하지 못한다.

원핫 인코딩은 단어사전 기준으로 벡터의 길이를 측정한다(TF IDF도 마찬가지)
> 그래서 전체 단어사전에 등장하는 단어가 1000개면, 벡터의 길이는 1,000이 된다. 

## TF-IDF

원핫 인코딩에 비해 **단어의 중요도**를 고려할 수 있는 방법이다.  

TF * IDF의 가중치를 모든 단어마다 계산하고, 원핫 인코딩의 1인 부분(즉,hot!
불이 켜진 부분)에 TF * IDF 값을 넣어준다.
> 이렇게 단어의 중요도를 반영했더니, 문장끼리의 유사성이 원핫인코딩 보다 잘 반영되더라!!

원핫 인코딩과 TF-IDF는 모두 **단어 사전(Vocabulary)**을 기준으로 벡터의 길이를 정한다는 점에서는 뿌리가 같습니다.

뼈대: 전체 데이터에 등장하는 단어가 1,000개라면, 벡터의 길이는 1,000이 됩니다.
(예: '경제'는 0번 방, '성장'은 1번 방...)

원핫 인코딩: 단어가 있으면 그 방에 1을 넣고, 없으면 0을 넣습니다. (불을 켜고 끄는 스위치 역할)

TF-IDF: 단어가 있으면 그 방에 단순히 1을 넣는 게 아니라, 계산된 **가중치(Weight)**를 넣습니다.
(스위치를 켜되, 얼마나 밝게 켤지 조절하는 다이얼 역할)

> 즉, 원핫 인코딩에서 0인 자리는 TF IDF 에서도 여전히 0이다.
> 반대로 단어가 존재해서 원핫 인코딩이 1이 되는 자리에만 가중치(TF IDF 값)이 채워지게 되는 것이다.


하지만, 여전히 원핫 인코딩의 문제점은 여전히 남아있다.
1) 희소 벡터의 문제가 있음  
2) TF IDF가 원핫 인코딩 보다는 정보를 더 많이 가지고 있기는 함..

## 임베딩(밀집벡터 만들기) - word2vec, Sentence BERT

*sparse 벡터 문제를 해결해보자! 동시에 TF IDF 보다 문장의 유사성을 더 잘 반영하는 방법을 고민해보자!!*

> TF-IDF는 단어의 문맥을 전혀 고려하지 못합니다. (Bag-of-Words 기반이기 때문)

> TF-IDF: "개가 사람을 물었다"와 "사람이 개를 물었다"는 등장하는 단어가 100% 일치하므로 **유사도 1.0(완벽 일치)**이 나온다. 

> 임베딩(특히 최신 모델): 단어 간의 관계를 고려하여 두 문장의 벡터를 다르게 형성할 수 있다. 

원핫 인코딩, TF IDF에서 존재했던 0을 없애고 이곳에 정보를 넣어 압축한 것이다.
이렇게 길게(vocab의 단어 수만큼 존재한 차원)이 압축되면서 밀집벡터가 된다

> 밀집(Dense): 원핫 인코딩처럼 대부분이 0인 '텅 빈 벡터'가 아니라, 모든 칸이 의미 있는 실수로 꽉 찬 벡터이기에 정보의 밀도가 매우 높다. 


### 임베딩(밀집벡터)의 철학: *분포가설*

"비슷한 맥락에서 쓰이는 단어는 비슷한 의미를 가진다." (NLP의 아주 강력한 가정임)  
> 우리는 '사과'라는 단어를 사전에서 찾아 그 뜻을 알기도 하지만, "사과가 달다", "사과를 먹다"와 같은 맥락을 통해 그 단어의 의미를 파악하기도 한다.
즉, '주변단어 곧 맥락'을 통해 그 의미를 알 수 있다.

> 인공지능도 코퍼스의 문장을 읽으며 '경제'라는 단어 주변에서 나오는 단어를 보며 자주 등장하는 패턴을 학습하는 것이다

> 그 결과 특정 단어 주변에 나타나는 이웃 단어들의 목록이 비슷하다? -> 그 단어는 유사한 단어다! 라고 판단하는 것이다.

> 기존의 단어들은 단어 그 자체의 철자 형태..
> 를 중시했다면, 임베딩은 그 단어의 주변 단어를 보고 그 정체를 파악하는 것이다!
> (사전적 의미에 집중하면 의미 하나에만 고정되어 언어적 유연성을 다룰 수 없다.)
> > 그 단어가 누구인지 알려거든 그 친구들을 보라!! 이런 의미

결국 임베딩의 가장 강력한 점은 *단어가 달라도 가까운 벡터*를 표현할 수 있다라는 것이다.
모델은 학습을 통해 형태는 다르지만, 비슷한 위치(주변단어)에 쓰인다는 것을알기 때문에 특정 문장을 벡터 공간의 가까운 곳에 배치한다.
> 사전에 적힌 뜻(단어의 형태, 철자))는 달라도 쓰임새는 비슷하네!!

### 임베딩을 하기 위해서는 *대량 텍스트에서 패턴 학습*이 필요하다

거대 데이터를 통째로 신경망에 집어 넣고, 문장 속의 단어를 가렸을 때 그 자리에 어떤 단어가 올지 맞추는 연습을 반복하면서 단어 간의 관계와 맥락을 파악하는 법을 정교하게 다듬는다(가중치 업데이트) 

> 이러한 점을 미루어 생각해 보면, 임베딩 벡터 수치는 어떤 코퍼스로 학습된 모델이냐에 따라서 다를 것이다.
> '경제'라는 단어를 똑같은 768차원 공간에 표현한다고 해도, 모델 A와 B의 학습 코퍼스가 다르다면?
> -> 좌표는 수치적으로도 위치적으로도 다를 것이다.


### 임베딩 행렬이란 무엇인가?

입데이 행렬은 **사전**이다. 

우리는 인코딩된 원핫 벡터를 임베딩 행렬이라는 사전을 통해 '밀집 벡터;로 변환한다. 

임베딩 행렬은 모델이 대량의 텍스트를 통해 **이미 학습을 마친 '가중치 테이블'**이다.
> 만약 임베딩 행렬이 (10.000, 768) shape의 행렬이라고 한다면?
> 행은 단어사전의 크기이기 때문에 vocab에 단어가 10,000개라는 의미이고.
열은 각 단어가 가질 특질의 갯수이다
>
> > 즉, 이 행렬의 각 행은 특정 단어(뉴스, 기사, 경제 등)가 가진 768차원의 의미 정보를 담고 있다!!
이런 식으로

#### 작동원리: 룩업

원핫 벡터(input)과 임베딩 행렬(weight)을 행렬곱 한다 -> 원핫 벡터에서 1이 있는 위치가 3번이라면, 행렬의 3번째 행만 남고 나머지는 0과 곱해지기 때문에 모두 사라진다.
-> 결국 1이 위치한 번호의 행을 테이블에서 그대로 꺼내오는 것과 같게 된다.
그러니까 실제 곱하는게 아니라 가져오는 것임!!
/ 임베딩 룩업 과정 -> 768차원의 밀집 벡터로 바뀐다.

이렇게 단어 하나하나 먼저 찾고, 그 결과등을 모아서 문장을 완성하는 식이다.
> 즉, 문장에 3개의 단어가 있다면, 컴퓨터는 임베딩 행렬로 달려가서 각 단어에 해당하는 768 차원의 벡터를 3개 꺼내온다.
>
> 우리는 문장 전체를 하나의 벡터로 만들고 싶기에..
단어별로 찾은 벡터를 하나로 합친다!(*이때 쓰이는 방법이 mean pooling - 더해서 평균내기 / 혹은 sentence Bert의 cls 토큰 사용*)

*계산 결과가 상상이 되는가? 선형대수학으로 생가해 보자*

원핫 행벡터: $(1 \times 10,000)$  (단어 한 개를 표현)

임베딩 행렬: $(10,000 \times 768)$ (단어 사전 전체의 의미 정보)

행렬곱 연산: $(1 \times \mathbf{10,000}) \cdot (\mathbf{10,000} \times 768) \rightarrow (1 \times 768)

결과값인 $(1, 768)$ 행렬은 사실상 768개의 숫자가 담긴 하나의 벡터와 같다. 

> 왜 "룩업(Lookup)"이라고 부르는지 다시 보기수학적으로는 행렬곱이지만, 실제 연산 결과를 뜯어보면 아주 재미있습니다.
> 원핫 벡터의 $n$번째 칸만 1이고 나머지는 다 0이기 때문에:0인 칸들은 임베딩 행렬의 해당 행들과 곱해져서 모두 0이 되어 사라집니다.
> 1인 칸만 임베딩 행렬의 n번째 행과 곱해져서 자기 자신 값을 그대로 남깁니다.

결국, **전체 행렬곱을 수행하는 것**과 **행렬에서 n번째 줄만 쏙 뽑아오는 것(Lookup)**은 수학적으로 완전히 동일한 결과를 낸다.
그래서 딥러닝 프레임워크(PyTorch 등)에서는 연산 속도를 위해 실제로 곱하기를 하지 않고 그냥 해당 번호의 행을 "꺼내오는" 방식을 사용하는 것이다.

위의 예시는 한 단어이기 때문에 문장으로 생각해 보면 

입력: $(10 \times 10,000)$ 행렬 (원핫 벡터 10개를 쌓은 것)

연산: $(10 \times \mathbf{10,000}) \cdot (\mathbf{10,000} \times 768) \rightarrow (10 \times 768)$

결과: 768차원짜리 단어 벡터 10개가 층층이 쌓인 $(10, 768)$ 행렬이 된다. 

이제 이 $(10, 768)$ 행렬을 앞에서 배운 것처럼 평균을 내거나 합치면, 드디어 우리가 목표로 했던 문장 전체를 대표하는 768차원 벡터가 탄생하는 것이다.
