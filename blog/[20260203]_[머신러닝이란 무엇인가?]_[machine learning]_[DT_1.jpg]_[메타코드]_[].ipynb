# 머신러닝이란 무엇인가?_머신러닝의 사상과 개념

---  
## 데이터사이언티스트에게 있어서 '선택'이란

데이터 사이언티스트로서 현실의 문제를 해결하기 위해 '선택'을 해야한다. 

여기서 선택이란?
> 어떤 알고리즘을 이용하여 어떤 모형을 만들 것인가?
> 어떻게 학습하고 제공할 것인가?(튜닝)  
> 어떻게 평가하고 이것을 활용할 것인가?  

결국, 기계가 다 생각해 주지 않는다. 어떤 알고리즘, 모형을 적용하는게 이 문제 해결에 적합할 지 그리고 어떤 HPO(하이퍼파라티머 튜닝)을 통해서 학습하는 것이 좋을지 또는 이걸 알려주는 알고리즘은 어떤 것을 선택할지... 사이언티스트가 선택해야한다. 

---

## 머신러닝이란  

가설검증이 연역적이라면, 머신러닝은 귀납적이다.

머신러닝의 핵심은 알고리즘에 기반한 모델 학습이다. (알고리즘과 모델 학습이 다르네~)  

머신러닝은 줄기세포와 같다. 줄기세포는 분화하여 심장, 간, 적혈구, 피부 등등... 이 될 수 있다.  
 
머신러닝을 주제로 생각해보면, 줄기세포는 '알고리즘'이다.  

알고리즘은 우리가 어떤 목적으로 가지고 어떻게 학습시키느냐에 따라서 분류, 예측, 군집, 추천, 이상탐지 등등의 모형으로 뻗어나갈 수 있다.  
> 우리가 알고리즘을 어떻게 학습시키는 지에 따라서 우리의 모델이 어떤 문제를 풀 수 있는 모델이 달라진다!!  
> 즉, 알고리즘 -> 모형 으로 학습!

이때, 사이언티스트는 알고리즘을 어떻게 학습시킬지 4가지 개념을 통해 정해줄 수 있다.

1. 목적함수(손실함수) Objective(Loss) function  
2. Hyper-Parameter(옵션)  
3. 학습, 검증, 테스트 셋 분리( Train, Validation, Test Sets) 
4. 평가 매트릭스(Evaluation Metrics)  

이 4가지를 이해하면, 우리가 가진 알고리즘을 학습하여 각각의 목적과 쓰임을 갖는 모델로 만들 수 있다. 

---

## 그 외  

물론, 어떤 알고리즘이냐에 따라서 그 한계가 결정되기는 한다. 하지만, 같은 알고리즘이더라도 어떻게 학습시키냐(예를 들어 HPO)에 따라서 모형의 결과와 성과 성능이 완전하게 달라지게 된다.  (알고리즘의 예시:SVM
K-NN
Decision-Tree
Artificial Neural
Diffusion )

머신러닝은 다음과 같은 특성이 있다.
1. 불확실성  
> 머신러닝 모델은 불확실성을 항상 가진다 - 어떻게 학습시키느냐에 따라서 달라지니까!

2. 적응성과 진화   
>학습시킨 모델 계속 학습할 수 있기 때문에 '적응성과 진화가 있다'

3. **블랙박스**
> 머신러닝 모델은 내부 동작 방식을 잘 알 수 없기에 블랙박스라는 특징을 가진다.  
> 결국 이러한 점을 극복하기 위해 등장한 것이 XAI이다.

4. 데이터의존성  
> 학습 방법이 아무리 좋아도 데이터가 좋지 않으면 좋은 모델이 나올 수 없다.   

5. 도메인 독립성  
> 가설검정에는 도메인이 필요하다.

하지만, 머신러닝은 데이터 안에 도메인의 특징이 녹아있기 때문에 '상대적'으로 도메인에 대한 독립성이 있다고 본다. 


BUT, 데이터사이언스와 사회과학을 하는 입장을 생각해 본다면, 도메인 지식은 오히려 매우 중요할 수 있다는 것을 나는 지속적으로 배워왔다.

### 알고리즘과 모형?(의사결정 나무로 생각해보면)

🔹 알고리즘 (Algorithm)

분할 기준 선택

불순도 계산 (Gini, Entropy)

노드 분할 반복

🔹 모형 (Model)

알고리즘을 통해 학습된 결과물

실제로 예측에 쓰이는 객체
(예: clf = DecisionTreeClassifier(...)

best_model = grid_search.best_estimator_)

| ---- | ------ |
| 알고리즘 | 요리법    |
| 모형   | 완성된 요리 |
| 데이터  | 재료     |
