# CNN 개념

## CNN은 왜 필요한가?

이미지 데이터의 특수성

문장은 시퀀셜 하지만, 이미지는 공간적이다

어느부분에 어떤 숫자가 있냐? 이게 중요함 즉, 공간적으로 접근해야 한다. 
(그러니까 이미지에 RNN하면 의미 없다)


일반 신경망으로는 이미지 데이터를 처리하기 부족했다. 

1. 파라미터 수 폭발 문제
> 28 by 28 이미지의 경우 784개의 뉴런이 필요하다. 224 224 3 이미지는? -> 150,528개의 노드가 필요하다! 그래서 파라미터 수가 폭발하고 이로 인한 과적합 우려가 있다.
>
> 
2. 공간적 관계 즉, 위치 정보가 무시됌
> 옆 픽셀, 위아래 픽셀의 관계가 소실된다. 즉, 공간적인 정보를 볼 수가 없다는 것이다. 






## CNN의 해결법

CNN은 일반신경망으로 이미지를 처리할 때의 문제(1. 파라미터 수 폭발 2. 공간 정보 반영 x)를 해결하기위해 **커널**로 이미지를 훑으며 지역적 특징을 추출하고, 같은 필터를 이미지 전체에 적용하여 파라미터를 공유함으로써(파라미터 수 대폭 감소) 이 문제를 해결했다.
> 이를 통해 파라미터 수 감소, 공간적 구조 보존이 가능했다. 

## 합성곱연산(Convolution) 


합성곱이 무엇이냐? -> 간단하게 말하면 **겹쳐서 곱하고 더한다**


합성곱이란, 작은 필터(커널)을 입력 위에서 슬라이드 시키면서 각 위치마다 곱하고 더해서 특징을 뽑아내는 연산이다. (-> 이를 통해 특징을 추출한다.)
> 합성곱 연산을 통해 이미지의 각 지역 정보를 재가공 할 수 있다. 
>

정리하자면, conv 연산이란? -> 커널을 이용해서 각 커널이 특정한 특징을 감지하는 것이다 

### 이미지 데이터의 구조

이미지 데이터의 각 픽셀은 0 ~ 255의 정수값을 갖는다. 
> 픽셀 안에 잇는 빛의 세기로 이미지를 숫자로 표현한다.  

이떄 숫자는 빛의 세기이다. 0 ~ 255 사이의 빛의 세기임. 까만색이면 0 255는 완전 하얀색

-> 이렇게 이미지도 숫자(행렬 또는 텐서)로 바뀐다 

## CNN을 위한 notation

커널(=필터): 학습 대상이다. 
> 임의로 커널을 만든다-> 커널 안에 있는 숫자가 w이다. 즉, 임의의 w를 준다

> 그리고 (3,3) 커널을 가장 위의 왼쪽에 붙여서 해당되는 셀을 곱하고 더하는 합성곱을 한다. 이걸 슬라이싱 하면서 반복!

이렇게 출력된 것이 피처 맵이다. 

왜 이런짓? -> 이미지의 정보를 재가공해서 전달해 주는 것이다. 그렇기 위해서 우리는 '압축'을 하는 것이다

특성 맵(feature map): 커널을 통해 출력된 결과 데이터이다. 각 커널이 다른 특징을 감지한다.
> 즉, 숫자 더미에서 그 성격을 대표하는 feature를 뽑는다(물론 뭔지는 모르고 모델이 알겠지) 그 성격을 대표하는 것들로 만든 map이다 -> 공간적 의미 잘 포착

스트라이드: 필터를 이동하는 간격
> 커널이 다음 커널로 갈 떄 셀 1만큼 이동하면 스트라이드 1이다

패딩: 입력 데이터 주변을 0으로 채우는 것이다. 이를 통해 출력 크기를 입력과 동일하게 유지가 가능하다. 
> 마지널한 영역에 0을 채워줌  

패딩하면 5,5가 7,7이 된다 

5,5를 합성곱 연산하면 3,3이 되는데
패딩을 입혀서 합성곱 연산하면 특성 맵이 5,5이 된다. 즉, 원래랑 shape 똑같아진다.
> 입력 데이터와 feature map의 shape를 동일하게 유지하면 뭐가 좋을까?  
> shape를 유지하지 않으면 피처 맵의 크기가 너무 빨리 작아져서 깊은 네트워크를 쌓기 어렵다. 또한, 동일하게 유지하는 것 외에도 가장자리 정보가 빨리 사라지는 것을 막기 위함도 있다. 

참고로 컬러 이미지의 경우 R G B 각 채널별 합성곱 결과를 더해서 하나의 특성 맵을 생성한다. 

수용영역: 특정 컨볼루션 레이어(또는 뉴런)의 출력 값 하나가 원본 입력 이미지의 몇 픽셀 영역으로부터 영향을 받는지(볼 수 있는지)를 나타내는 공간적 범위

결국 CNN은 수용영역 내 공간패턴을 기억한다

## CNN의 구조

우리는 큰 이미지를 벡터화해서 마지막에 고양이 강아지로 분류해야 한다. 

마지막  1000차원의 벡터를 가지고 강아지 고양이 분류는 쉬운데

224,224 벡터 이런걸로 분류하기는 어렵다. 
그래서 우리는 공간의 정보를 차곡차곡 연산해서 줄여서 벡터로 피는 것이다. 



224 224 가 112 112 로 줄어드는 것을 풀링이라고 한다. (pooling이라는 말은 여러개의 숫자를 어떻게 해서든지 하나로 만든다! 합친다! 이런 뜻이다) 

그게 어떤 식이든 풀링은 큰 면적을 작은 면적으로!



## CNN 전체 구조 흐름

1. **입력 이미지** (224, 224, 3) (컬러 이미지의 3차원 텐서 구조)
   - 분석의 대상이 되는 원본 이미지 데이터

2. **합성곱(Conv) + ReLU** [특징 추출]
   - 커널(Filter)을 이용해 이미지의 선, 면 등 주요 특징을 추출하고 비선형성 추가

3. **풀링(Pool)** [크기 축소]
   - 중요 정보를 유지하면서 데이터의 공간적 크기를 줄여 연산량 감소

4. **Conv + Pool 반복** [깊은 특징]
   - 위 과정을 반복하며 사물의 형체나 추상적인 깊은 특징을 학습

5. **Flatten** [1차원 변환]
   - 추출된 다차원 특징 맵을 신경망(FC)에 넣기 위해 1열로 나열

6. **FC층 + Softmax** [분류 결과]
   - 모든 특징을 종합하여 최종 결과(예: 강아지 90%, 고양이 10%)를 도출


*참고: (224,224,3)이냐 (3,224,224)냐의 차이는 텐서플로우 방식이냐 파이토치 방식이냐의 차이! 둘 다 똑같다.*

전체 흐름을 봤으니 하나하나 까보자...


입력 이미지가 들어오면, 합성곱 연산(커널로 겹치고 곱하고 더하고 슬라이싱..)을 적용하고 ReLu 함수를 거친다. -> 이 과정이 이미지의 특징을 추출하는 과정이다

> 합성곱 연산이 컨볼루션 레이어 Conv layer에서 일어난다. 이 레이어에서는 이미지에서 직접적인 특징을 추출한다. 그리고 각 레이어는 활성화 맵을 생성한다.   
> 이 계산 결과는 ReLu 활성화 함수를 사용하는데, 비선형 특성을 효과적으로 학습 가능하다.

그 다음 **풀링(pooling)**을 거친다 (pooling layer에서)  
앞서 언급한 것처럼 풀링은 줄이는것! 풀링을 왜 하냐? 한마디로 정보의 핵심만 남기고 덩치를 줄이는 것이다.  
> 풀링 레이어에서 특징 맵의 차원을 축소하여 계산 비용을 줄이고, 모델의 지엽적이고 사소한 노이즈를 제거하여 과적합을 방지한다. 풀링을 통해 불변성을 확보할 수가 있다(사물의 위치가 조금 바뀌어도 동일한 사물로 인식할 수 있도록 특정 영역에서 강한 신호만 뽑아내는 것이다)

결국 풀링이란 불필요한 정보는 버리고 요약하는 것이다. 

이 과정을 반복한다!! -> 그 결과 층이 깊어질 수록 데이터의 가로 세로는 작아지지만(풀링해서) 찾아낸 특징의 종류(채널 수)는 많아진다. 

이때, conv layer의 역할은 위치에 따라 조금 달라진다.
> 입력층에 가까운 conv layer 즉, 하위 레이어에서는 이미지의 가장 낮은 수준의 특징을 감지한다 선 각도 색상 변화 등..  
> 출력층에 가까운 상위 레이어에서는 더 복잡하고 추상적인 특징을 학습한다. 또한, 하위 레이어에서 추출된 특징을 조합하여 특징을 통합한다. 


Flatten [1차원 변환]

마지막 분류를 위해서는 줄을 세워야 한다!!   
7 by 7 by 512처럼 입체적으로 쌓여 있던 특징들을 한 줄로 길게 펴서 1차원 벡터로 만든다. 


FC층 + Softmax [분류 결과]  

마지막으로 추출된 모든 특징들을 종합해서 결론을 내리는 부분이다.  

> FC층 (Fully Connected): 한 줄로 세운 특징들을 최종 후보군(예: 강아지, 고양이, 자동차)과 연결합니다. 각 특징이 어떤 클래스에 얼마나 기여하는지 가중치를 곱해 점수를 냅니다.

> Softmax: 나온 점수들을 다 더하면 **100%(1.0)**가 되도록 확률로 변환합니다.

>> 예: [강아지: 0.92, 고양이: 0.05, 자동차: 0.03] → "이 사진은 92% 확률로 강아지입니다!"

**conv + ReLu & pooling을 반복할 수록 데이터의 가로 세로는 짧아지지만, 층은 쌓인다. 이거 왜 그럴까? 어떻게 이해해야 할까?**


**가로·세로가 줄어드는 것은 '공간 정보의 압축'**이고, **층(채널)이 두꺼워지는 것은 '의미 정보의 확장'**이다.  

여기서 층은 채널이다!!
> channels란 입력 이미지에서 특징 맵의 깊이를 의미한다.  
> 채널의 크기는 필터 수에 의해 결정되는 것이다!

만약, conv layer를 지났는데 채널이 1에서 20으로 늘어났다고 가정해보자(이 부분이 CNN의 핵심 마법 중에 하나이다) -> 이 과정은 곧 서로 다른 20종류의 커널이 각각 이미지를 훑어서 서로 다른 20개의 feature map을 만든 것이다.

1. 필터의 개수가 곧 출력 채널의 개수이다. 입력 채널이 1이라는 것은 마치 흑백 이미지와 같다. 이걸 conv layer를 지나 출력 채널 20을 만들고 싶다면? -> 우리는 서로 다른 숫자가 적힌(즉, 서로 다른 커널 w를 가진) 20개의 필터를 준비한다.
> 필터 1은 세로선을 잘 찾고... 필터 2는 대각선을 잘 찾고.. 이런 식으로

2. 1번 필터가 이미지를 처음부터 끝까지 훑으며 합성공 수행 -> 1번 커널의 특징 지도(이것은 1채널)이 나온다.

3. 이 작업을 20번 반복한다

4. 그 결과 20개의 특징 지도가 생겼다. 이렇게 각각 다른 w를 가진 커널에서 만들어진 feature map을 쌓는다!

5. 이렇게 채널이 20개인 하나의 데이터가 만들어진다!!

그럼 왜 채널을 늘려야 하는가? 1채널의 단순한 이미지가 20개의 필터를 거치면? *20개의 서로 다른 관점으로 해석한 결과물*이 나온다!


1. 만약 풀링이 없다면? (계산 폭탄)

VGG-16 모델을 예

 입력 이미지는 $224 \times 224$
 
 풀링을 전혀 안 했을 때: 마지막 특징 맵(Feature Map)의 크기가 여전히 224 224이고 채널이 512라면, Flatten을 했을 때 벡터의 길이는 무려 25,690,112가 됩니다.
 
 문제점: 이 엄청난 길이를 다음 FC층(완전 연결층)에 연결하려면 수십억 개의 가중치(W)가 필요합니다. 메모리가 버티지 못하고 터져버리겠죠.
 
 2. 풀링이 들어갔을 때 (다이어트 성공)VGG-16은 중간중간 5번의 풀링을 거쳐 가로·세로를 7 7까지 줄입니다.  
 3. 풀링 후: 7 7 512를 Flatten 하면 벡터 길이는 25,088입니다.결과: 풀링이 없을 때보다 데이터 크기가 약 1,000배나 줄어들었습니다! 이제야 비로소 일반적인 컴퓨터로 계산 가능한 수준이 됩니다  


"정보 손실"보다 "핵심 요약"

"길이를 줄이기 위해" 풀링을 하지만, 단순히 자르는 게 아니라 **가장 강한 신호(Max)**만 남기며 줄이는 것이 핵심이다. 

Flatten 전의 상태: "이 픽셀은 강아지 코고, 저 픽셀은 0.1mm 옆에 있는 강아지 코야." (너무 세세함)

풀링 후의 상태: "어쨌든 이 영역(7x7) 안에는 강아지 코가 확실히 있어." (핵심 요약)

이렇게 공간적인 사소한 차이는 무시하고 "무엇이 있는가"라는 정보만 남겨서 Flatten으로 넘겨주기 때문에, 모델은 훨씬 효율적으로 분류를 수행할 수 있게 됩니다.

주의점

풀링은 단순 기계작업이다. 즉, 업데이트 (학습)의 대상이 아니다. 학습 대상은 커널의 w이다.
> 특성 맵의 크기를 줄이는 연산인 풀링은 중요한 특징을 유지하면서 불필요한 세부 부분을 제거하는 것이다.


헷갈리는 부분

1. channel이 무엇인가?

컬러 이미지는 각 픽셀을 RGB 3개의 실수로 표혀한 3차원 데이터이다. 그러므로 컬러 이미지는 3개의 채널로 구성된다. 만약, 흑백 명암 이미지면? 2차원 데이터로 1개 채널로 구성된다. 

만약 컬러 이미지의 경우처럼 입력 데이터가 여러 채널을 가지는 경우 필터는 각 채널을 순회하며 합성곱을 계산한 후, 채널 별 feature map을 만든다. 그리고 각 채널별 map을 합산하여 최종 feature map으로 반환한다. 즉, 입력 데이터는 채널 수와 상관없이 필터 별로 1개의 feature map이 만들어진다. 

**conv layer에 유입되는 입력 데이터는 한 개 이상의 커널이 적용할 수 있다. 이 경우에 feature map에는 필터 갯수만큼의 채널이 만들어진다. 즉, conv layer에 n개의 필터가 적용된다면, 출력 데이터는 n개의 채널을 갖게 된다. 


## CNN 단계 별 레이어 구분


특징 추출 단계 (합성곱 연산과 pooling을 반복하여 이미지의 feature를 추출하는 단계)
- convolutin layer: 필터를 통해 이미지의 특징을 추출하는 레이어  
- pooling layer: 특징을 강화시키고 이미지의 크기를 줄이는 레이어

이미지 분류 단계(classification)
- flatten layer: 데이터 타입을 FC 네트워크 형태로 변형함. 입력 데이터의 shape 변경만 수행하는 layer
- softmax layer: classification 수행
