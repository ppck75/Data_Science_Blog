# XAI(설명 가능한 인공지능)


---

## XAI란 무엇인가?

인공지능 모델의 내부 작동원리를 사람이 해석 가능한 형태로 제공하여
모델의 동작방식을 파악할 수 있게 하는 것!

---

## 왜 필요해?

우리는 의사결정 나무 알고리즘을 통해 학습한 모델이 어떻게 학습되었는지 이미지화 할 수 있었다.

그런데, 다른 여러 모형들은 이렇게 모델의 구조를 시각화하기 어렵다. 못한다.


인공지능의 필연적 한계점 **'블랙박스 모형과 불투명성'**이 있다. 이게 필연적인 한계점이다. 인공지능 모델이라는 것은 우리가 직접 만들었음에도 불구하고 어떤 동작 매커니즘으로 어떻게 나왔는지 인간이 해석 가능한 형태로 확인하는 것이 굉장히 어렵다.


    물론, 머신러닝의 경로나 과정을 확인할 수 있다. 하지만, 일종의 로그를 우리가 찍어낼 수 있어도 그걸 인간이 해석할 수는 없다. 
    
    이런 한계점을 가지고 있기 때문에 머신러닝의 결과를 현업에 사용하려고 하면, 담당자는 우리는 그걸 왜 그렇게 나왔는지 확인을 해주셔야지 우리가 이걸 사용할 수 있어요... 라는 물음에 답하기 어렵다. 

그래서 우리는 이러한 것들을 대안적으로 해소하기 위해 XAI라는 개념을 도입했다. 

XAI는 인공지능 모델의 내부작동원리를 사람이 해석가능한 형태로 제공하여 모델동작방식을 파악할 수 있게 하는 것이다. 


사람이 해석 가능한 형태 -> 이걸 위해서 보통 세가지 기법을 사용한다. 

---


## 인공지능의 블랙박스 모형과 불투명성

우리가 말하는 블랙박스 모형은 단순히

“복잡해서 이해하기 어렵다”
는 수준이 아니야.

보다 정확히 말하면,

입력 → 출력의 함수는 명확히 정의되어 있지만,
그 내부 추론 구조가 인간의 인지 체계로는 압축·요약 불가능한 상태

를 의미한다. 

### 왜 블랙박스는 ‘필연적’인가?

이건 기술의 한계라기보다 성능 추구의 결과야.

🔹 (1) 성능과 해석 가능성의 근본적 트레이드오프

단순한 모델
→ 구조 명확
→ 설명 가능
→ 성능 제한적

복잡한 모델
→ 비선형성, 고차원 상호작용
→ 성능 우수
→ 해석 불가

즉,

블랙박스성은 실패가 아니라,
고성능 모델이 치르는 구조적 비용이다.

> 로짓모형을 생각해보아라... 기존 단순 선형 회귀보다 특정 상황에서 예측력과 설명력이 더 좋다. 하지만, 그 내부 원리를 이해하기 점점 힘들어진다. 
이게 왜 ‘현업’에서 문제가 되나?

이건 단순히 학술적 문제가 아니야.

 1) 책임 소재 문제

현업에서는 항상 이 질문이 따라와:

“이 결과가 잘못되면 누가 책임지죠?”

모델? ❌

개발자?

담당자?

👉 설명 불가능한 모델은 책임 귀속이 불가능하다.


2) 규제와 법적 요구


금융,
의료,
채용,
공공 행정 등 많은 영역에서 이미: 

“왜 이런 결정을 내렸는지 설명할 의무”

를 요구받고 있다. 

이때 블랙박스 모델은:

성능이 좋아도 ❌

설명 불가 → 사용 불가!



> 결론적으로 XAI는 내부 작동 원리 즉, 블랙박스로 보이지 않는 그 불투명성을 단순화, 근사 , 요약해서 해석가능하도록 전달하는 것이다.
> 그러므로 XAI는 블랙박스를 제거하지 않는다. 블랙박스를 해석 가능한 수준으로 만드는 것일 뿐이다(그렇다고 완전한 해석도 아니다) 

### 정리 
인공지능 모델의 블랙박스성은 기술적 실패라기보다 고성능 추론이 필연적으로 동반하는 구조적 특성이다. 현대의 머신러닝 모델은 고차원 비선형 공간에서 학습되며, 그 내부 추론 과정은 형식적으로는 투명하더라도 의미적으로는 인간이 이해하기 어려운 형태를 띤다. 이러한 불투명성은 특히 실제 현업 적용에서 책임성과 신뢰성의 문제로 이어지며, 모델의 성능과 무관하게 사용을 제한하는 요인이 된다. XAI는 이러한 한계를 제거하기보다는, 모델의 판단을 인간이 해석 가능한 수준으로 근사·요약하여 제공함으로써 블랙박스의 위험을 관리하려는 시도라고 볼 수 있다.

## XAI의 세가지 기법

1. Feature Importance -> 모델’에 대한 각 변수의 기여도를 본다. (단, 학습할 때의 기여도이다)   
    각 변수의 모델 학습 시 등수를 세운다. 

    우리는 이 기여도를 보고, 이 모델이 이러한 변수가 가장 크게 작용하여 이러한 분류를 했습니다~~ 라고 말할 수 있다. 

2.  Partial Dependence Plot -> 변수의‘구간별’기여도

3.  Shapley value -> ‘예측’에 대한 각 변수의 기여도 (1번이 학습에 대한 각 변수의 기여도라면, 이것은 예측에 대한 각 변수의 기여도를 의미한다.)
