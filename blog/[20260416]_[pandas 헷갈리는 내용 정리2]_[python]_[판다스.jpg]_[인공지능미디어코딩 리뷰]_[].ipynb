## 데이터 프레임의 특정 부분 선택하기 (selcetion)


데이터프레임에서 원하는 행과 열을 골라내는 방법

| 방법 | 기준 | 예시 |
|------|------|------|
| `df['열이름']` | 열 이름 | `mydata['사고건수']` |
| `df[['열1','열2']]` | 여러 열 이름 | `mydata[['사고건수','사상자수']]` |   -> 데이터프레임으로 나오겠지
| `df.iloc[행, 열]` | **위치(숫자)** 기준 | `mydata.iloc[0, 2]` → 1행 3열 |
| `df.loc[행, 열]` | **이름(레이블)** 기준 | `mydata.loc[0, '사고건수']` |

loc[0, '열이름']: "이름표가 **0**이라고 써진 행에서 '열이름' 칸을 보여줘." (이름 기준)

iloc[0, 2]: "데이터의 맨 첫 번째(0번째) 행에서 세 번째(2번째) 열을 보여줘." (순서 기준)

> 근데 loc 할 때, 행은 보통 숫자로된 인덱스인 경우가 대부분이고, 열은 보통 글자로된 변수명을 갖기 때문에 loc(0,'사고건수') 이런 식의 경우가 많다!

```python
# '이름' 컬럼을 행의 이름(인덱스)으로 설정했다고 가정
# 인덱스: ['홍길동', '김철수', '이영희']
# 컬럼: ['나이', '사고건수']

# 1. loc 사용 (이름 기준)
mydata.loc['홍길동', '사고건수']  # 가능! (이름표를 보고 찾아감)
mydata.loc[0, '사고건수']         # 에러! (0이라는 이름표를 가진 행이 없음)

# 2. iloc 사용 (순서 기준)
mydata.iloc[0, 1]                # 가능! (이름이 뭐든 첫 번째 줄의 두 번째 칸)
```

### loc와 iloc를 쓰는 이유?




df['열'], df[['열1','열2']]와 결정적으로 다른 차이는
**행과 열을 동시에 잡을 수 있는가?** 이다.

df['열이름']은 오직 열만 선택할 수 있다. 행을 선택하려면 df['열이름'][0] 처럼 두 번 작업해야 한다

loc와 iloc는 콤마 하나로 어떤 행의 어떤 열을 가져올 지 결정할 수 있다.


또한, 슬라이싱이 강력하다. 


df['열이름']은 여러 열을 가져올 때 df[['열1', '열2']]처럼 일일이 이름을 다 써줘야 합니다. 하지만 loc은 **"여기부터 저기까지"**가 가능하다.

df.loc[0:5, '사고건수':'사상자수']: 0~5번 행까지, 그리고 '사고건수'부터 '사상자수' 사이에 있는 모든 열을 다 가져와!

이건 일반 인덱싱으로는 불가능한 기능이다. 


또한, 조건문에서 많이 활용! (특정 조건을 만족하는 행의 특정 열만 가져와라 이런 식으로 )

> df.loc[df['나이'] > 20, '이름']
>
> 만약, 인덱싱으로만 쓴다면 df[df['나이'] > 20]['이름'] 이런식으로 사용해야 한다.
