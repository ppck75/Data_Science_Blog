````python
# 인덱싱 (key로 value 접근)

dct = {"한국": "서울", "미국": "워싱턴", "중국": "베이징"}
print(f"한국의 수도: {dct['한국']}")
print(f"일본의 수도: {dct.get('일본', '데이터 없음')}")  # 없는 key → 기본값 반환
print(f"일본의 수도: {dct.get('일본')}")

#  dct.get(key, default)는 딕셔너리에서 키 값을 안전하게 가져오는 함수입니다.

#   - 키가 있으면 그 값을 반환
#   - 키가 없으면 default를 반환 (default를 안 주면 None)

#   dct["c"]는 키가 없으면 KeyError가 나지만, get()은 에러 없이 기본값을 줄 수 있다는 점이 핵심 차이입니다.
````

참고

get() 메서드는 딕셔너리에서 주어진 키(key)에 해당하는 값을 반환하는 메서드입니다.

get() 메서드는 두 개의 파라미터를 가질 수 있습니다. 첫 번째 파라미터는 찾고자하는 키(key)이고, 두 번째 파라미터는 기본값(default value)입니다.

get() 메서드는 다음과 같은 동작을 수행합니다:

만약 주어진 키(key)가 딕셔너리에 존재한다면, 해당 키에 대한 값을 반환합니다.
만약 주어진 키(key)가 딕셔너리에 존재하지 않는다면, 기본값(default value)을 반환합니다. 만약 기본값이 지정되지 않았다면, None을 반환합니

````python
# 중첩 for문
for i in range(2, 10):       # 2단 ~ 9단
    for j in range(1, 10):   # ×1 ~ ×9
        print(f"{i} × {j} = {i*j}")
    print()                  # 단 사이 빈 줄
````

````python
# 리스트 컴프리헨션에서 else
numbers = list(range(1, 16))   # [1, 2, ..., 15]

# 2단 만들기 (9 미만만)
result = [f"{i} × 2 = {i*2}" for i in numbers if i < 10]
print(result)

# if-else 포함 (else는 for 앞에 위치, 그리고 [] 안에)
result2 = [f"{i} × 2 = {i*2}" if i < 10 else "범위초과" for i in numbers]
print(result2)

# for 뒤에 if와, for 앞의 if는 어떻게 다른가?
# 리스트 컴프리헨션에서 for 뒤에 오는 if는 “필터(조건 통과한 요소만 포함)”이고, for 앞에 오는 if ... else는 “각 요소에 대해 값을 고르는 조건식”이다.
#즉, for 뒤 if - 필터링 담당 / for 앞 if - 값 선택(조건식)

# numbers에서 요소 하나 꺼내고 i<10이면? f"{i} × 2 = {i*2}"를 리스트에 담는다. 만약, 거짓이면 범위초과를 리스트에 담는다. A if 조건 else B -> 조건이 True A실행, 아니면 B실행이다. 
````

# lambda 함수

한 줄짜리 간단한 함수를 **즉석에서** 만드는 문법

```python
lambda 매개변수: 표현식
```

````python
# lambda 함수


인사 = lambda x: f"안녕하세요. {x}입니다." 

# 이렇게 하면 그냥 함수가 되어버린다.!! 
# # 읽다가 람다 나오면 아 한 줄 함수구나!
print(인사("BTS"))

mean = lambda *args: sum(args) / len(args) # 람다 함수로 mean을 만든거지
print(f"평균: {mean(7, 9, 11)}")

# 한줄짜리 함수 이름을 쓸 때에는 '함수이름 = lambda 매개변수: 표현식' -> 이것도 형식을 기억하자. 
````

````python
# 합격 불합격을 예측하는 함수를 한줄짜리 람다 함수로 만들어보자

합불여부판독 = lambda score: '합격' if score > 90 else '불합격'
print(합불여부판독(85))
````

````python
class Calculation:                # 클래스는 앞에 대문자로 쓰는게 관행, 함수는 보통 소문자로 하는 것이 관행
    def __init__(self, n1, n2):   # 생성자: 두 숫자를 입력받아 속성으로 저장 # __init__ 구간에서 속성을 정의해준다. 나머지는 다 함수이다. 
        self.n1 = n1              # 속성 1 눈 = 파란색
        self.n2 = n2              # 속성 2 코 = 들창코, 작은코(n2로는 이런게 들어온다)

    def add(self):                # 메서드: 더하기 (_init__ 아닌 모든 것은 다 함수이다)
        return self.n1 + self.n2

    def sub(self):                # 메서드: 빼기
        return self.n1 - self.n2

    def mul(self):                # 메서드: 곱하기
        return self.n1 * self.n2

    def div(self):                # 메서드: 나누기
        return self.n1 / self.n2

# 객체 생성 (클래스는 대문자, 객체는 소문자 관례)
cal = Calculation(8, 5) # 속성 설정을 먼저 해줘야 하니까 (8,5)를 넣어준다. n1이 8 n2가 5인 Cal 속성 가지고 있는 인스턴스 cal이 생성된다. 
````

````python
print(cal.n1, cal.n2)

# self.n1 self.n2 -> 8,5를 속성을 갖는 cal 인스턴스 설정 -> cal.n1 cal.n2가 된다. 
````


클래스는 붕어빵 틀이고, 객체(인스턴스)는 그 틀에서 찍어낸 붕어빵입니다. 틀을 만들 때는 나중에 나올 붕어빵의 이름을 미리 알 수 없기 때문에, 임시 이름표인 self를 붙여두는 것입니다.

설계도(Class): "앞으로 만들어질 **나(self)**의 n1에 값을 넣어줘."

실제 객체(Instance): cal = Calculation(8, 5)라고 선언하는 순간, 내부적으로는 self 자리에 cal이라는 이름이 대입된다고 보시면 됩니다.

1. 내부에서 일어나는 일
질문하신 cal.n1을 호출할 때, 컴퓨터 내부에서는 대략 이런 흐름이 일어납니다.

##### 📘 Python 인스턴스와 self의 동작 원리

| 단계 | 코드상 표현 | 실제 동작 (개념적 이해) |
| :--- | :--- | :--- |
| **인스턴스 생성** | `cal = Calculation(8, 5)` | `Calculation.__init__(cal, 8, 5)` 호출 |
| **속성(값) 저장** | `self.n1 = 8` | `cal.n1 = 8` (cal이라는 방에 8을 저장) |
| **메서드 호출** | `cal.add()` | `Calculation.add(cal)` (cal의 데이터를 사용해 더하기) |

3. 왜 굳이 self라고 쓸까?
하나의 클래스로 여러 개의 객체를 만들 수 있기 때문입니다.

Python
cal1 = Calculation(10, 2)
cal2 = Calculation(50, 5)
이때 cal1.add()를 하면 self는 cal1이 되고, cal2.add()를 하면 self는 cal2가 됩니다. 덕분에 똑같은 add 함수를 써도 각자 자기의 숫자들을 찾아가서 계산할 수 있는 것이죠.


 self는 **"인스턴스가 생성된 후의 자기 이름"**이 들어갈 자리라고 이해하시면 완벽합니다. 덕분에 cal.n1처럼 외부에서 접근할 때나, 클래스 내부 함수들끼리 데이터를 공유할 때 헷갈리지 않고 자기 데이터를 찾아갈 수 있습니다.


# 파일 읽기와 쓰기(with, open...)

`open()` 함수로 파일을 열고, `with` 문을 사용하면 **자동으로 파일을 닫아줌**

| 모드 | 설명 |
|------|------|
| `"w"` | 쓰기 (기존 내용 덮어쓰기) |
| `"r"` | 읽기 |
| `"a"` | 추가 (기존 내용 뒤에 이어쓰기) |

파이썬 하다보면 한글, 영어 여러 파일들을 읽고 저장하고 불러올 수 있어야 한다 (py나 ipynb말고도)

with는 무엇을 열어 작업하고 닫아! 까지 할 때 사용한다. 즉, close를 쓰기 귀찮은거지. open하고 작업하고 바로 자동 close 를 하고 싶을 때 with을 쓴다 

형식이 중요함
> with open('경로','모드', 인코딩) as f:  (열어서 f라고하자)  
    (들여쓰기)  f.write('쓸 내용')                    (f에다가 써)

````python
with open("test.txt", "w", encoding="utf-8") as 문서: 
    문서.write("RM, 진, 슈가, 제이홉, 지민, 뷔, 정국") # 모드가 w니까 write. read()는 에러가 난다
print("✅ test.txt 파일 저장 완료")

# f.write()는 with문에 포함된 것이다
#   위에서 f.write(...)만 with 안에 있고, print(...)는 들여쓰기가 없으므로 with 밖에서 실행됩니다.
#   즉, 파일 쓰기가 끝나고(자동으로 닫힌 뒤) print가 실행됩니다.
````
