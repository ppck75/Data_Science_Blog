# Numpy 

Numpy는 기본적으로 배열에 연산을 하면 배열을 돌려준다 이렇게 설계되어 있다. 하지만, 결과가 스칼라로 나오기도 한다. 즉, 전체를 하나로 줄이는 연산이면 array가 아닌 숫자 하나 (스칼라)가 나올 때도 있다. 혹은 shape관련 함수는 종종 tuple로 준다. 

즉, 대부분 array에 함수 적용 -> array

## Numpy의 인덱싱과 슬라이싱

  arr2d = np.array([
      [10, 20, 30],
      [40, 50, 60],
      [70, 80, 90]
  ])

  | 표현 | 의미 | 결과 |
  |---|---|---|
  | arr2d[1] | 1행 전체 | [40 50 60] |
  | arr2d[:, 1] | 1열 전체 | [20 50 80] |
  | arr2d[1, 2] | 1행 2열 값 하나 | 60 |
  | arr2d[[0, 2]] | 0행, 2행 선택 | [[10 20 30], [70 80 90]] |
  | arr2d[:, [1, 2]] | 1열, 2열 선택 | [[20 30], [50 60], [80 90]] |
  | arr2d[[0, 2], [1, 2]] | 좌표쌍 (0,1), (2,2) 선택 | [20 90] |


팬시 인덱싱은 행 인덱스를 리스트로 준 경우 열 인덱스를 리스트로 준 경우 모두 다 해당

즉, 인덱스 자리에 리스트나 배열을 넣어서 여러 위치를 고르는게 팬시 인덱싱이다. 

다른 표로 정리하자면, 

  | 표현 | 한 줄 요약 |
  |---|---|
  | arr2d[1,2] | 값 하나 |
  | arr2d[[1,2]] | 행 여러 개 |
  | arr2d[:, [1,2]] | 열 여러 개 |
  | arr2d[[1,2],[0,2]] | 좌표 여러 개 |

  - arr2d[1] -> 일반 인덱싱
  - arr2d[0:2] -> 슬라이싱
  - arr2d[[0, 2]] -> 팬시 인덱싱
  - arr2d[:, [1, 2]] -> 팬시 인덱싱

## reshape(-1)

reshape(행,열) -> 지정한 shape로 변환하는 함수이다

그런데 -1을 주면? 
- 전체 원소 수는 이미 정해져 있으니
- 다른 차원 수를 보고
- 남는 크기를 NumPy가 자동 계산

arr = np.array([1,2,3,4,5,6])

  arr.reshape(2, -1)

  여기서 원소는 6개이고, 앞 차원을 2로 정했으니
  뒤 차원은 자동으로 3이 되어:

  array([[1, 2, 3],
         [4, 5, 6]])

  즉 reshape(2, -1) 는 reshape(2, 3) 과 같아진다. (즉, 자동계산 해준다 -1로 두면!)

 - reshape(-1) -> 1차원으로 펼침  
  - reshape(2, -1) -> 2행으로 만들고 열 개수는 자동 계산  
  - reshape(-1, 3) -> 3열로 만들고 행 개수는 자동 계산
  - -1의 뜻 -> 남은 차원 크기를 자동으로 계산  

*reshape(-1)은 보통 배열을 1차원으로 펼친다!라는 뜻으로 많이 쓰인다*

  - arr.flatten() : 1차원으로 펼침
  - arr.ravel() : 1차원으로 펼침
  - arr.reshape(-1) : 1차원으로 펼침

## Numpy의 브로드캐스팅
*예시로 두 조건을 이해해보자*

브로드캐스팅 규칙

1. 두 축의 길이가 같거나
2. 둘 중 하나의 축의 차원의 크기가 1일 때

조건1 예시

```
예시 2: (3, 1) + (1, 3) → (3, 3)

  A = [[1],    b = [[10, 20, 30]]
       [2],
       [3]]

  A + b = [[11, 21, 31],
           [12, 22, 32],
           [13, 23, 33]]
```



조건2 예시 

```
예시 1: (2, 3) + (3,) → (2, 3) + (1, 3) → (2, 3)

  A = [[1, 2, 3],     b = [10, 20, 30]
       [4, 5, 6]]        ↑ 자동으로 2행으로 확장

  A + b = [[11, 22, 33],
           [14, 25, 36]]
```

  | shape A | shape B | 가능 여부 | 이유 |
  |---|---|---|---|
  | (3,) | (3,) | 가능 | 길이가 같음 |
  | (2,3) | (2,3) | 가능 | 각 축 길이가 같음 |
  | (2,3) | (1,3) | 가능 | 한 축은 1, 다른 축은 같음 |
  | (2,3) | (2,1) | 가능 | 한 축은 1, 다른 축은 같음 |
  | (3,) | (2,) | 불가능 | 길이 다르고 1도 아님 |

Numpy는 축을 맞출 때에는 앞에 1을 붙여서 (3,)을 (1,3)으로 보고 브로드태스팅 해준다. 그런데, 이것은 둘 중 하나가 1이어야만 한다. (3,) (2,)를 보면, 둘 다 1차원이라 안된다. (1,3) (1,2)로 봐준다고 하더라도 뒤에 차원이 안맞게 되기 때문에 브로드캐스팅 적용 못한다. 




### 차원의 짝이 맞으면 브로드캐스팅 가능하다?

“차원의 짝이 맞는다” = shape가 완전히 같다 는 뜻은 아니고, 비교하는 각 축이 브로드캐스팅
  규칙에 맞게 대응된다는 뜻에 더 가깝습니다.

  핵심은 오른쪽 축부터 비교하는 겁니다.

  예를 들어 shape가 완전히 같은 경우:

  (2, 3)  vs  (2, 3)

  이건 당연히 됩니다.
  이 경우는 네 말대로 shape가 같은 경우예요.

  그런데 shape가 같지 않아도 되는 경우가 있습니다.

  (2, 3)  vs  (3,)

  NumPy는 (3,) 를 오른쪽에 맞춰서 비교합니다:

  (2, 3)
  (1, 3)

  이제 축별로 보면:

  - 3 vs 3 -> 같음
  - 2 vs 1 -> 하나가 1

  그래서 가능합니다.

  즉, shape는 다르지만 차원의 짝은 맞는 것입니다.

  또 예를 들면:

  (4, 1, 3)  vs  (   5, 3)

  NumPy는 짧은 쪽 앞에 1을 채워서 봅니다:

  (4, 1, 3)
  (1, 5, 3)

  축별 비교:

  - 3 vs 3 -> 같음
  - 1 vs 5 -> 하나가 1
  - 4 vs 1 -> 하나가 1

  그래서 이것도 가능합니다.

  반면 이건 안 됩니다:

  (2, 3)  vs  (2,)

  오른쪽부터 맞추면:

  (2, 3)
  (1, 2)

  축별 비교:

  - 3 vs 2 -> 다름, 둘 다 1도 아님

  그래서 불가능합니다.

  정리하면:

  - shape가 같으면 당연히 가능
  - 하지만 shape가 달라도, 오른쪽부터 비교해서
      - 같거나
      - 둘 중 하나가 1
        이면 가능

  즉,

  “차원의 짝이 맞는다”는 말은 shape가 완전히 같다는 뜻이 아니라, 각 축이 브로드캐스팅 규칙대
  로 대응 가능하다는 뜻입니다.

  짧게 표로 보면:

  | shape A | shape B | 가능? | 이유 |
  |---|---|---|---|
  | (2,3) | (2,3) | 가능 | 완전히 같음 |
  | (2,3) | (3,) | 가능 | 오른쪽부터 맞추면 (1,3) |
  | (4,1,3) | (5,3) | 가능 | (1,5,3)로 맞춰 비교 가능 |
  | (2,3) | (2,) | 불가 | 3 vs 2 충돌 |

  한 줄 요약:

  shape가 같은 것은 “차원의 짝이 맞는” 경우의 한 예일 뿐이고, 브로드캐스팅에서는 shape가 완
  전히 같지 않아도 축별로 호환되면 연산할 수 있습니다.
