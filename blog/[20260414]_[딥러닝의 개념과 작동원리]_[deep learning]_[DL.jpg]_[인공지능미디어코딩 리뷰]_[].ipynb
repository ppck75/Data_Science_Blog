# 딥러닝의 개념과 작동 원리

## 퍼셉트론

퍼셉트론이란 초기 인경 신경망으로써, 뉴런에서 신호를 받아 임계치를 넘으면 다음 뉴런에 전달하는 뇌의 뉴런 구조를 모방했다. 
> 입력값 x의 가중치의 합이 임계치를 넘으면 1을 출력하는 방식

수학 구조: 
z = w1*x1 + w2*x2 + ... + b (선형 변환)  
y = f(z) (활성화 함수적용. 이 활성화 함수 부분이 있기에 input의 가중치의 합이 특정 임계치가 넘으면 1을 출력할 수 있음) 

-> 하지만, XOR 문제를 풀 수 없었다. (직선 하나로 풀 수 없는 문제)

## 딥러닝의 탄생 (MLP)

XOR 문제를 MLP로 해결함!!

입력층과 출력층 사이에 (활성화 함수를 포함한) 은닉층을 추가하여 비선형 문제를 해결했다. 은닉층이 2개 이상부터 DNN 라고 부른다.  
> 입력층이 input 데이터를 받고 출력층이 ouput을 출력하는 층이라고 한다면, 은닉층(H)는 특징을 추출하는 층이다. 
>
> *전결합층 (FC layer, Dense layer): 모든 입력 노드가 다음 층의 모든 출력 노드와 하나하나 빠짐없이 연결된 층*


이때, 입력층 -> 은닉층 -> 출력층 방향으로 진행되는 연산 구조를 FFNN(피드포워드 신경망)이라고 한다. 

> 역전파 알고리즘을 사용해도 FFNN이다. FFNN은 모델이 결과물을 내놓는 추론 과정이 순방향이라는 의미니까! 입력 -> 은닉 -> 출력 방향으로 '추론'이 이뤄지고, 뒤로 돌아오지 않는 특성 때문에 FFNN이라고 부른다. 이런 개념을 생각해보면 RNN는 순환하기 때문에 FFNN과 반대이다. 

## 히든 레이어

XOR 문제로 다시 돌아가보자,

'선형의 늪'이 있다.  
> 선형 + 선형 = 선형  
> 직선 함수(선형)을 아무리 겹치고 쌓아도 하나의 선형 함수가 될 뿐이다. 1번 레이어와 2번 레이어를 아무리 합쳐도 선형으로만 표현될 뿐이다. 이것은 단 한개의 층을 가진 커다란 퍼셉트론이랑 똑같을 것이다 -> XOR 문제 풀 수 없음

여기서 등장한 것이 비선형성을 부여하는 활서오하 함수이다.  
> 즉, 히든 레이어가 데이터를 다른 차원으로 보내면 그 차원을 구부리는 것이 활성화함수이다. 

**즉, 히든레이어 + 활성화 함수**가 XOR 문제를 해결했던 것이다!(히든 레이어만 추가했다고 이 문제가 해결된게 아니다. 활성화함수도 있다. )

> XOR 문제의 해결은 단순히 층을 더 많이 쌓아서 해결된 것이 아니라, 그 층들이 활성화 함수가 적용된 층이었기 때문이다. 활성화 함수 없이 층을 100층 쌓아도 결과는 선형 결합일 것이다. 
>

### 활성화함수 종류

1. 시그모이드 함수: 이진분류 출력층에 주로 사용 / 문제점: 은닉층에 사용하면 기울기 소실 문제가 있다. 출력이 0이나 1에 가까워지면 미분값이 거의 0이 되기 때문에!  
> 딥러닝의 학습 과정은 역전파 과정에서 이뤄지는데, 이때 연쇄 법칙에 의해 각 층의 기울기(즉, 미분값)를 계속 곱하게 된다.  
> 그런데, 시그모이드의 미분값은 최대가 0.25이다. 이런 작은 값들이 여러 층에서 곱해지면서 결국 앞쪽 층으로 갈수록 기울기는 사실상 0이 된다 -> 학습 일어나지 않음. 

2. 소프트맥스: 다중 클래스 분류의 출력층에 주로 사용함. / 출력층에서 각 클래스의 확률 합은 1이며, 가장 높은 확률을 가진 클래스로 예측을 진행함

3. 렐루: 음수 -> 0 양수 -> 그대로 반환한다. / 은닉층에서 주로 사용한다. -> 기울기 소실 문제를 완화하기 때문에!   
> max(0, x) -> 둘 중 더 큰 값을 선택해라! 즉, x랑 0 가져와바, 그 둘 중 더 큰거를 f(X)로 내보낼게!  
> 이 방식으로 양수는 그대로 반환하고 음수는 0으로 반환할 수 있게 된다. 

### 기울기 소실 문제에 대해 좀 더 생각해 보자

시그모이드를 히든 레이어의 활성화 함수로 사용하는 3층짜리 신경망을 가정해보자. 
 

$$\frac{\partial Loss}{\partial w_1} = \frac{\partial Loss}{\partial y} \cdot \frac{\partial y}{\partial h_2} \cdot \frac{\partial h_2}{\partial h_1} \cdot \frac{\partial h_1}{\partial w_1}$$

위의 식은 w1을 업데이트 하기 위해 loss를 w1으로 미분한 것이다.
> 모델의 파라미터가 100만개 있으면, 위와 같은 식이 100만개 있는 것이다. 물론, 이걸 하나하나 계산하는게 아닌, 행렬로 묶어서 한번에 처리한다.  

$$W = \begin{bmatrix} w_1 \\ w_2 \\ \vdots \\ w_{10^6} \end{bmatrix}, \quad \nabla W = \begin{bmatrix} \frac{\partial L}{\partial w_1} \\ \frac{\partial L}{\partial w_2} \\ \vdots \\ \frac{\partial L}{\partial w_{10^6}} \end{bmatrix}$$

이렇게 묶으면 100만 개의 식이 단 한 줄의 행렬식으로 변합니다.$$W_{new} = W_{old} - \eta \cdot \nabla W$$

다시, 돌아와서 w1을 업데이트 하기 위해 미분한 값을 연쇄법칙으로 쪼개보면, 각 층을 통과할 때마다 활성화 함수의 미분값 (시그모이드의 미분값)이 계속 곱해지는 형태이다. 


시그모이드 함수를 미분한 값은 최댓값이 0.25이다. 이렇게 층이 곱해진다면? 층이 10개만 되어도 0.25가 10번 곱해지는 것이다. 거의 0에 수렴하게 된다

$\frac{\partial L}{\partial w_1}$   이 0에 수렴하게 된다!!

그럼 가중치 w1은 미분값이 0이기 때문에 아래의 공식에서 Wnew와 Wold의 값이 똑같게 된다. 즉, 학습이 일어나지 않는다 

$$W_{new} = W_{old} - \eta \cdot \nabla W$$

$$W_{new} \approx W_{old}$$

아래로 볼록한 2차 함수 모양의 손실함수가 있다고 할 때, 이 그래프에서 기울기가 가파르다는 것은 현재 위치가 최적점(골짜기 바닥)에서 멀리 떨어져 있음을 의미한다.

그러므로, 특정 w를 미분했을 때 미분값이 크면? -> 많이 업데이트를 해야한다!


grad가 큰 w에 대해서는 오차를 줄이기 위한 많은 수정이 필요할 것이다(즉, 가중치가 크겠지)

*역전파 (Backward): 오차를 뒤로 전달하며 각 가중치(Weight)가 오차에 기여한 정도(Gradient)를 계산합니다.*
> 즉, 미분값이 크면? -> 가중치가 오차에 기여한 정도가 크다 -> 수정 많이 필요!
