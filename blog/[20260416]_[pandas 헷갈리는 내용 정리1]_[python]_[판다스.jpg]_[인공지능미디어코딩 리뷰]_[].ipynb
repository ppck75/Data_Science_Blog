### 핵심 자료구조 3가지

pandas가 자료를 이해하는 구조

| 자료구조 | 설명 | 비유 |
|---------|------|------|
| **DataFrame** | 행(row)과 열(column)로 이루어진 **2차원 표** | 엑셀 시트 전체 |
| **Series** | DataFrame에서 **열 1개**를 뽑아낸 **1차원 데이터** | 엑셀 시트의 열 하나 |
| **Index** | 각 행을 구별하는 **레이블(번호)** | 엑셀 시트의 맨 왼쪽 행번호 |

````
       Index    열1(Series)   열2(Series)
       ─────   ───────────   ───────────
 0  →  0       100           '서울'
 1  →  1       200           '부산'       ← 이 전체가 DataFrame
 2  →  2       300           '대구'
````

시리즈: 변수 하나에 딸린 자료들 쭉~ (즉, 열로 보는 것이 시리즈)  
인덱스: 맨 왼쪽에 잇는 열 ID를 인덱스라고 부른다. 

데이터프레임, 시리즈는 자료형이 '판다스 인스턴스'이다. 즉, 판다스라는 라이브러리(안에 이쓴 클래스)로부터 만들어진 클래스 객체 즉, 인스턴스이다.
> 그러므로 데이터프레임, 시리즈, 판다스의 인덱스.. 이건 자료형이 판다스 인스턴스이다. 

> 참고로 컬럼 이름 -> 변수

## 데이터 탐색하기

| 함수/속성 | 반환값 | 용도 |
|----------|--------|------|
| `.info()` | 변수명, 데이터 타입, 결측치 | 전체 구조 파악 |
| `.head(n)` | 상위 n행 | 앞부분 미리보기 |

## 변수(열) 삭제 (분석에 불필요한 정보를 삭제하자!)


> 분석이 깔끔해진다 단, 원본을 보존해야 한다면 inplace = False로 하고 새 변수에 할당하는 방법을 쓰자

`````python
# 분석에 불필요한 변수 삭제
# axis=1: 열(변수) 방향, inplace=True: 원본 직접 수정
mydata.drop(['데이터기준일자', '제공기관코드', '제공기관명'], axis=1, inplace=True) 


# axis=1 은 열을 봐라! 축 지정 안하면 오류난다!!!
# inplace=True 안쓰면 False가 default이다.

mydata.drop('사고지역관리번호',axis = 1) 
# 만약 drop하고 싶은 칼럼이 하나라면, 굳이 리스트로 안하고 이런 식으로! 
# 지우려는 대상이 하나일 때는 그냥 문자열만 쓰고, 여러개일 때는 리스트에 담아서 주고
`````

data.drop('칼럼이름')으로 특정 열은 지웠는데, 그럼 행을 삭제하고 싶을 떄는?
> 중복이 있을 수도 있다

`````python
# 행 삭제 및 중복 제거 (참고)
# ──────────────────────────────────────────────

# 특정 인덱스의 행 삭제 (예: 인덱스 0, 1, 2번 행)
mydata.drop([0, 1, 2], axis=0, inplace=True)  # axis=0: 행 방향을 지정해준다!

# 중복 행 삭제
mydata.drop_duplicates(inplace=True)                       # 모든 변수 기준 # 중복 행 삭제하는 함수이다. 
mydata.drop_duplicates(subset=['사고지역위치명'], inplace=True)  # 특정 변수 기준
`````

mydata.drop() : "내가 지정한 특정 번호를 지워줘"
> 오직 인덱스만 보고 동작한다.  

mydata.drop_duplicates() : "내용이 똑같은 행이 있으면 하나만 남겨줘"  
> 내용이 100% 똑같은 데이터가 두 번 들어왔다면 하나만 남기고 나머지는 중복이니까 삭제해! 
> > subset 옵션: subset=['사고지역위치명']이라고 쓰면, 다른 칸은 달라도 '사고지역위치명'이 똑같은 데이터가 있으면 중복으로 간주하고 지운다.   

즉, subset 옵션을 쓰지 않으면 모든 칼럼 값이 100% 일치해야 중복으로 판단하고 제거한다. 단, 하나의 칼럼 값이라도 다르면 이건 다른 데이터로 생각하고 버리지 않는 것이다.

그런데 subset 옵션에 칼럼 이름을 주면, "다른 칼럼 값 상관없고 이 칼럼의 값만 똑같으면 하나만 남겨!" 가 되는 것이다. 지정한 컬럼의 값이 같다면, 나머지 컬럼의 값이 서로 달라도 중복으로 간주하고 하나만 남긴 채 나머지는 삭제합니다.

`````python
# Series 추출 및 속성 확인
# ──────────────────────────────────────────────

# DataFrame에서 '사고건수' 열을 Series로 추출
mydata_건수 = mydata['사고건수']

# Series 주요 속성
print(f"타입: {type(mydata_건수)}")          # pandas.core.series.Series
print(f"이름: {mydata_건수.name}")            # 사고건수
print(f"크기: {mydata_건수.shape}")           # (12780,) — 1차원
print(f"데이터 타입: {mydata_건수.dtype}")      # int64

# 처음 5개 값 확인
print(f"\n처음 5개 값:\n{mydata_건수.head()}")
`````
