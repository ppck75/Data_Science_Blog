문제 1) while loop / break 응용

1부터 어떤 수까지 차례대로 더할 때, 그 합이 5000을 처음으로 초과하는 순간의 숫자를 구하고자 한다.

그 숫자를 [inum]
그때까지의 누적합을 [isum]

이라는 이름으로 저장하시오.
단, isum은 반드시 5000을 초과한 값이어야 한다.

```python
inum = 0
isum = 0

while isum <= 5000 : 
    inum = inum + 1
    isum = isum + inum
print(isum, inum)
```

문제 2) 조건이 있는 for loop

1부터 100까지의 숫자 중에서, 30 이상 40 이하의 숫자는 제외하고 모두 더한 값을 구하시오.

결과를 [res] 라는 이름으로 저장하시오.

단, 반복문 안에서 조건문을 사용하시오.

```python
res = 0
for x in range(1,101):
    if (x >= 30) and (x <= 40):
        continue 
    res += x
print(res, x)
```

문제 3) dataframe 생성과 merge

다음 두 표를 각각 pandas dataframe으로 만드시오.

```python
import pandas as pd
```

```python
dtA = pd.DataFrame({'id':[1,2,3],'socre2024':[80,75,90]})
dtB = pd.DataFrame({'id':[1,3,4],'socre2025':[85,95,70]})
print(dtA)
print(dtB)
```

```python
# merge 해보자

dt_inner = pd.merge(dtA,dtB, on = 'id', how ='inner')
print(dt_inner)

dt_left = pd.merge(dtA, dtB, on = 'id', how = 'left')
print(dt_left)

dt_right = pd.merge(dtA, dtB, on = 'id', how = 'right')
print(dt_right)

dt_outer = pd.merge(dtA, dtB, on = 'id' , how = 'outer')
print(dt_outer)
```

문제 5) wide-form을 long-form으로 변환

다음 표를 [wdt1] 이라는 이름의 pandas dataframe으로 만드시오. 

```python
wdt1 = pd.DataFrame({'id': [1,2,3],'wage_2021':[10,20,15],'wage_2022': [12,18,17]})
```

```python
wdt1
```

**[wdt1]**를 reshape하여 long-form으로 만들되,
2021과 2022의 값은 year라는 변수로 만드시오.

long-form을 id 변수, year 변수 순서대로, 작은 값이 위로 오게(ascending) sort하여
**[ldt1]**이라는 이름으로 저장하시오.

**[ldt1]**의 1-3열은 id, year, wage 순서대로 있어야 합니다.

```python
ldt1 = pd.wide_to_long(wdt1,['wage'],i='id',j='year',sep='_')
ldt1
```

```python
ldt1.columns
```

```python
ldt1 = ldt1.reset_index()
```

```python
ldt1
```

```python
ldt1.sort_values(by = ['id','year'],inplace= True)
```

```python
ldt1 = ldt1.reset_index().drop('index',axis= 1)
ldt1
```

문제 6) long-form을 wide-form으로 변환

다음 표를 [ldt4] 라는 이름의 pandas dataframe으로 만드시오. 첫 번째 행은 변수명입니다.



```python
ldt4 = pd.DataFrame({'pid':[1,1,2,2,3,3], 'year':[2021,2022,2021,2022,2021,2022],'score':[80,85,75,78,90,88]})
ldt4
```

**[ldt4]**를 reshape하여 wide-form으로 만들고,
결과를 **[wdt4]**라는 이름으로 저장하시오.

최종 결과는 각 사례가 한 행이 되도록 하며,
변수명은 pid, score2021, score2022 순서가 되도록 하시오.

```python
wdt4 = pd.pivot(ldt4, index = 'pid',columns = 'year')
wdt4
```

```python
wdt4 = wdt4.reset_index()
```

```python
wdt4.columns = [f'{a}{b}' for a,b in wdt4.columns]
wdt4
```
