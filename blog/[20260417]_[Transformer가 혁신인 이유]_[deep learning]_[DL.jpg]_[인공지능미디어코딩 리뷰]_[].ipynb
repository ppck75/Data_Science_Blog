# Transformer가 혁신인 이유

seq2seq 모델

인코더, 디코더 구조를 가졌고 입력 시퀀스를 하나의 고정 길이 벡터인 context vector로 압축해서 (입력 문장의 정보를 압축) 디코더에 전달한다.

디코더는 context vector를 받아 출력 시퀀슬 생성한다.


이때 seq2seq 모델의 한계가 있는데, 입력 문장이 아무리 길어도 고정된 길이의 context vector에 압축해야 하기 떄문에 정보 손실이 불가피 했으며 병목 현상이 있었다.

**이것을 attention 매커니즘을 도입해서 해결하고자 했다**

attention: 디코더가 출력을 만들 때마다 인코더의 모든 은닉 상태를 참조하여 관련 부분에 주목한다. 이로 인해 context vector에만 의지하지 않게 된 것이다.

기존 Seq2Seq는 인코더의 마지막 결과물인 고정 크기 컨텍스트 벡터 하나만 디코더에게 넘겨줬지만, 어텐션이 적용된 버전은 인코더의 **모든 시점의 히든 스테이트($h_1, h_2, \dots, h_n$)**를 통째로 디코더에게 공유한다 

  Self-Attention은 단어 간 관계를 계산하는 메커니즘이고, Multi-Head Attention은 이를 여
  러 관점에서 동시에 수행하는 방식이며, 이러한 구조 덕분에 Transformer는 입력 단어들을
  병렬적으로 처리할 수 있다.
  - Self-Attention: 단어들끼리 서로 참고하는 방법
  - Multi-Head Attention: 그 참고를 여러 관점에서 동시에 하는 방법
  - 병렬 처리: 이 구조 덕분에 단어들을 한 번에 계산할 수 있는 성질


그럼 병렬처리는 왜 될 수 있었을까? 
> RNN은 각 단어를 계산할 때 이전 단어의 hidden state가 있어야 했다. 그래서 병렬 처리하기 어렵다. 하지만, 트랜스포머와 self attention에서는 각 단어에 대해서 Q, K , V를 만드는데 이것을 만들 때 이전 단어의 계산 결과를 기다릴 필요가 없다. 그래서 문장 전체를 한 번에 행렬 연산으로 처리할 수 있다. 

결국 트랜스포머가 그렇게 난리가 난 이유는 
기존의 모든 제약(RNN)을 완전히 부수고 어텐션만으로 '완벽한 시스템'을 구축했기 때문입니다

트랜스포머 이전의 어텐션은 항상 RNN이라는 엔진에 의존하는 '보조 장치'였습니다. 하지만 RNN은 단어를 하나씩 순서대로 읽어야만 하는 순차적 처리의 한계가 있었죠.

혁신: 트랜스포머는 RNN을 아예 삭제했습니다.  
효과: 문장이 아무리 길어도 모든 단어 사이의 관계를 한 번에(Parallel) 계산할 수 있게 되었습니다. 이는 GPU의 연산 능력을 100% 활용할 수 있게 하여, 학습 속도를 수십, 수백 배 앞당겼습니다. 지금의 GPT나 BERT 같은 거대 모델은 RNN 기반이었다면 학습에만 수십 년이 걸렸을 겁니다.

'거리'의 개념을 파괴함 (Constant Path Length)  
RNN은 아무리 어텐션을 써도, 정보를 전달할 때 옆 단어로 계속 넘겨줘야 하는 구조적 부담이 있었습니다. 문장이 길어지면 앞쪽 정보가 희미해지는 '장기 의존성 문제'에서 완전히 자유롭지 못했죠.

혁신: 트랜스포머의 Self-Attention은 첫 번째 단어와 1,000번째 단어를 직접 연결합니다.

효과: 두 단어 사이의 거리가 얼마나 멀든 상관없이 단 한 번의 연산으로 소통합니다. "정보가 전달되다 사라질(기울기 소실)" 걱정 자체가 사라진 것입니다.

'다각도'로 문장을 보는 시각 (Multi-Head Attention)  
이전의 어텐션은 보통 하나의 관점으로만 문장을 훑었습니다. 하지만 트랜스포머는 어텐션을 여러 개(Head)로 쪼개서 동시에 돌립니다.

혁신: Multi-Head Attention을 통해 같은 문장이라도 어떤 헤드는 '문법'을 보고, 어떤 헤드는 '의미'를 보고, 어떤 헤드는 '인칭 대명사'를 봅니다.

효과: 문장을 인간처럼 입체적으로 이해할 수 있게 되었습니다.

## 정리

트랜스포머가 혁신인 이유

  1. Self-Attention                                                                     
     각 단어가 문장 안의 다른 모든 단어를 직접 참고할 수 있게 만들었습니다.             
     이게 Transformer의 가장 핵심적인 변화예요. RNN처럼 먼 거리 정보를 단계적으로 전달하지 않고, 단어 간 관계를 직접 계산합니다.(장기 의존성 문제 완화)                                         
  2. Multi-Head Attention                                                               
     어텐션을 한 번만 하지 않고 여러 head로 나눠서 서로 다른 관계를 동시에 보게 합니다. 
     예를 들어 어떤 head는 주어-동사 관계를, 다른 head는 수식 관계를 더 잘 포착할 수 있 습니다. 즉, self-attention의 표현력을 확장한 장치예요.                             
  3. 병렬처리 구조                                                                      
     RNN처럼 이전 시점 결과가 있어야 다음 시점을 계산하는 구조가 아니기 때문에, 입력 전 체를 한꺼번에 계산할 수 있습니다.                                                  
     그래서 학습 속도가 크게 빨라졌고, 대규모 데이터 학습에 훨씬 유리해졌습니다.


 - 가장 본질적인 혁신: Self-Attention                                                  
  - 성능을 더 끌어올린 핵심 장치: Multi-Head Attention                                  
  - 구조적으로 얻어진 큰 장점: 병렬처리
